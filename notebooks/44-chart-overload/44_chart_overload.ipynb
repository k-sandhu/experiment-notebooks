# When a model is given more than one chart can show, what does it leave out?

## A chart for a ten-second look

Someone at a meal-kit company has to prepare one chart for Thursday's pricing meeting. In April the company raised the price of its Family plan from \$79 to \$89 a week, and the committee will decide whether to reverse the rise. The analyst has a table of monthly figures for all four plans and a page of notes from working through it. The chart will be on the screen for about ten seconds before the discussion starts.

The committee needs one fact above all: since the price went up, the share of Family customers who cancel each month has doubled, from about 3% to about 6.5%. It also needs a warning. The September figures were pulled halfway through the month, so September's low cancellation rate covers half a month and is not a recovery. A chart that shows September without saying so invites the committee to conclude that the problem has gone away.

The analyst's notes hold plenty more, and all of it is true. The Couple plan is the largest. The Vegetarian plan is growing fastest and has the best ratings. Support tickets peaked in December, sign-ups peak in January, and gift subscriptions are not counted. None of this bears on whether to reverse the Family price rise. Making the chart is mostly an editing job: choosing what to leave out without leaving out what the decision depends on.

## What goes wrong

People now often hand this job to an AI model. They paste in the table and the notes, ask for a chart, and get back code that draws it. A common complaint about the results is that they say too much. Every point gets a value label, every plan gets a line and a legend entry, a second panel appears for a second measure, and a footnote adds three more facts. Each label is accurate. Together they leave the reader to do the editing the chart was meant to do, in a meeting where nobody has time for it.

The opposite failure exists too. A chart can be clean and still drop the one warning that changes the decision, or show the right trend without a number anyone can repeat. A useful chart does both jobs: it carries little enough that the main message is obvious at a glance, and it carries everything critical.

### An illustrative example

Here is a simplified, made-up version. Suppose the table held only the four plans' monthly cancellation rates and the analyst wrote three notes: (a) Family cancellations doubled after the price rise; (b) September is half a month; (c) the Vegetarian plan is growing fastest.

One way to chart this draws four coloured lines and prints every monthly value beside its point: 48 numbers, a legend with four entries, and note (c) as a footnote because it seemed interesting. The reader's eye goes to whichever line is highest or brightest. Note (b) is missing, so September's dip looks like good news.

Another way draws the Family line in red and the other three plans as one grey average line. The title states the finding, "Family-plan cancellations doubled after the April price rise". Two numbers are printed, the average rate before and after the rise, and a short note sits beside September. This chart has far fewer marks than the first and carries both (a) and (b). It leaves out (c), which is true but does not affect the decision.

A person who makes the second chart starts from the decision, picks the message, and then asks of every line, label and colour whether it helps the committee decide. The question is whether a model does the same.

### How much to tell the model

There are two broad ways to ask. One is to hand over the material and say "make a chart". The model then has to work out for itself who the chart is for, what they will decide and which notes matter. The other is to tell it more. You can say who the audience is and what they must decide. You can add general design advice: state the finding in the title, use one strong colour for what matters, and print only the numbers that carry the message. You can go further and say which notes are critical and which to leave off, which does the editing for the model. Or you can show the model the chart it drew and ask it to improve it, as a designer looks at a draft.

Each step costs the person asking a little more effort, and the last two need either knowledge of what matters or an extra round. If a bare request already gives clean charts, the extra instruction is wasted. If clean charts only come when someone spells out the priorities, the model is not doing the editing and the person still has to.

## The question

A model receives a data table and analyst notes that hold more than one chart should show. Does it choose the critical information, show it, and leave the rest off? How do the number of labels on its charts, and the critical information a reader can take from them, change as the instructions grow from a bare request, to the purpose, to design principles, to explicit priorities, and finally to a look at its own rendered chart?

## Study summary

The study uses twelve made-up briefs like the meal-kit one, each a data table plus ten analyst notes: two critical facts, one critical caveat and seven true notes the decision does not need. Three models made one chart for every brief at each of five levels of instruction, 180 charts in all, each written as matplotlib code: Claude Opus 5.5, Muse Wonder Full and the open-weight DeepSeek V4.1 Flash. Every chart was measured three ways. The harness counted the labels and other text drawn on it. A reader model that saw only the image was asked for the main message, the caveat and the critical numbers. Three judge models that saw the brief rated the chart and marked which notes it called out: Claude Opus 5.5, Muse Wonder Full and GPT-6.1 Sol, whose maker has no chart in the study.

- With only "make a chart from this", Muse Wonder Full put 24 labels on an average chart and Claude Opus 5.5 put 16. DeepSeek V4.1 Flash put 6, but used legends and extra panels instead. The reference charts drawn for this study carry 4. Saying who the chart is for and what they decide removed few labels: 12 on average, and DeepSeek rose to 10.
- Six general design rules (title states the finding, one highlight colour, label only what carries the message, and so on) cut the labels to 5 per chart. They also cut the critical numbers. The reader could read only 57% of them, against 84% with the bare request, because 32% were no longer on the chart.
- When the prompt also quoted the three critical notes and asked for the rest to be left off, the reader read 96% of the critical numbers and found the main message on every chart. These charts carried 7 labels, 2 more than with the general rules alone, and almost no secondary notes.
- Shown the chart it had drawn, a model kept roughly the same content (5.6 labels, 60% of critical numbers) but cleaned up the drawing: overlapping text and garbled dollar amounts nearly disappeared, and the judges listed 0.4 errors per chart instead of 0.9.

Back in the meal-kit meeting: a bare request gives the committee a chart with everything on it, so the critical facts are there but have to be found among a dozen or more labels. General design advice gives a clean chart that often leaves off the number the committee needs, such as the cancellation rate before and after the rise. Clean and complete charts came only when the person asking said which notes mattered, which means the person still did the editing.

The caveat that most affects this result is the reader. It is a model that reads every label however small, and it found the main message on all 36 charts made from bare requests, including one with 118 labels. A person glancing at a slide for ten seconds is likely to be hurt more by clutter than this reader was; the judges' clarity rating (3.7 out of 5 for bare requests, 4.5 with design rules) points the same way. Every code reply rendered (178 first time, 2 after one repair), so format failures play no part in the comparison.

## 2. The twelve briefs

Each brief is a small, fictional business or public-sector situation, generated in code for this study. Every brief has the same parts:

- an audience and the decision it faces, such as "the pricing committee, deciding whether to reverse the April price rise on the Family plan";
- a data table with more columns and rows than one chart needs (24 to 220 rows, 7 to 12 columns);
- ten analyst notes, each one sentence or two, with numbers computed from the table.

Of the ten notes, two are critical facts that bear on the decision, one is a critical caveat that changes how the data must be read, and seven are true but secondary. Some secondary notes are tempting on purpose: the highest bar, the fastest-growing group, a record day. The models never see these roles; they see the notes in a fixed order that mixes the three kinds.

The data is synthetic because the test needs exact answers. Every number in every note and every quiz answer is computed from the same table by the code below, so a reader's answer can be checked to a tolerance, and no model can have seen the table before. The generator is seeded and its output is frozen in `data/briefs/`; a re-run checks that the regenerated files match.

The twelve situations cover the usual shapes of business charts: a trend with a break (B01, B08), a comparison of groups (B02, B04, B12), an A/B test (B03), a distribution (B05), a relationship between two measures (B06), a change in a ranking (B07), survey results (B09), cohort retention (B10) and shares across channels (B11).

````python
# Twelve fictional briefs, generated in code so every number in the analyst notes and every quiz answer
# is computed from the same table. Each finding has a role: "critical" (bears on the decision),
# "caveat" (changes how the data must be read) or "secondary" (true, but not needed for the decision).

def _pct(x, n=1):
    return f"{x:.{n}f}%"


def _money(x, n=1):
    return f"${x:,.{n}f}"


def _finish(bid, slug, topic, audience, decision, df, columns, findings, quiz):
    assert list(df.columns) == list(columns), (bid, list(df.columns))
    out = []
    for i, (role, text) in enumerate(findings, 1):
        assert role in ("critical", "caveat", "secondary")
        out.append({"id": f"F{i}", "role": role, "text": text})
    roles = [f["role"] for f in out]
    assert roles.count("critical") == 2 and roles.count("caveat") == 1 and roles.count("secondary") == 7, bid
    return {"id": bid, "slug": slug, "topic": topic, "audience": audience, "decision": decision,
            "df": df, "columns": columns, "findings": out, "quiz": quiz}


def brief_b01():
    rng = np.random.default_rng(4401)
    months = pd.period_range("2025-10", "2026-09", freq="M")
    rise = pd.Period("2026-04", "M")
    spec = {"Single": (8000, 260, 2.4, 45, 4.3), "Couple": (21000, 700, 3.0, 65, 4.4),
            "Family": (14000, 470, 3.0, 79, 4.2), "Vegetarian": (5000, 330, 2.7, 62, 4.6)}
    rows = []
    for plan, (start, base_signup, base_churn, price, rating) in spec.items():
        active = start
        for m in months:
            after = plan == "Family" and m >= rise
            season = {1: 1.6, 12: 0.8}.get(m.month, 1.0)
            frac = 14 / 30 if m == months[-1] else 1.0
            signups = round(rng.normal(base_signup * season * (0.65 if after else 1.0), base_signup * 0.04) * frac)
            churn = ((5.9 if m == rise else 6.6) if after else base_churn) + rng.normal(0, 0.12)
            cancels = round(active * churn / 100 * frac)
            tickets = round(active * 0.018 * (1.9 if m.month == 12 else 1.0) * frac * rng.uniform(0.93, 1.07))
            rows.append({"month": str(m), "plan": plan, "price_per_week_usd": price + (10 if after else 0),
                         "active_subscribers": active, "new_signups": signups, "cancellations": cancels,
                         "cancellation_rate_pct": round(100 * cancels / active, 2),
                         "avg_rating": round(rating - (0.15 if after else 0) + rng.normal(0, 0.03), 2),
                         "support_tickets": tickets})
            active = active + signups - cancels
    df = pd.DataFrame(rows)
    fam = df[df.plan == "Family"].set_index("month")
    pre = fam.loc[[str(m) for m in months if m < rise], "cancellation_rate_pct"].mean()
    post = fam.loc[[str(m) for m in months if rise <= m < months[-1]], "cancellation_rate_pct"].mean()
    a_apr, a_sep = fam.loc["2026-04", "active_subscribers"], fam.loc["2026-09", "active_subscribers"]
    sep_rate = fam.loc["2026-09", "cancellation_rate_pct"]
    first, last = df[df.month == "2025-10"].set_index("plan"), df[df.month == "2026-09"].set_index("plan")
    growth = (last.active_subscribers / first.active_subscribers - 1) * 100
    assert growth.idxmax() == "Vegetarian" and growth.drop("Family").min() > 0 and growth["Family"] < 0
    assert last.active_subscribers.idxmax() == "Couple"
    full = df[df.month != "2026-09"]
    by_month = full.groupby("month")[["support_tickets", "new_signups"]].sum()
    assert by_month.support_tickets.idxmax() == "2025-12" and by_month.new_signups.idxmax() == "2026-01"
    churn_by_plan = full.groupby("plan").cancellation_rate_pct.mean()
    assert churn_by_plan.drop("Family").idxmin() == "Single"
    rating = df.groupby("plan").avg_rating.mean()
    assert rating.idxmax() == "Vegetarian"
    columns = {
        "month": "calendar month (YYYY-MM)", "plan": "subscription plan",
        "price_per_week_usd": "weekly price of the plan, US dollars",
        "active_subscribers": "active subscribers at the start of the month",
        "new_signups": "new subscriptions started during the month",
        "cancellations": "subscriptions cancelled during the month",
        "cancellation_rate_pct": "cancellations as a percentage of active_subscribers",
        "avg_rating": "average customer rating of delivered boxes, 1 to 5",
        "support_tickets": "customer support tickets opened during the month"}
    findings = [
        ("secondary", f"Couple is the largest plan, with {last.active_subscribers['Couple']:,} active subscribers at the start of September 2026."),
        ("secondary", f"Vegetarian is the fastest-growing plan: active subscribers rose {growth['Vegetarian']:.0f}% between October 2025 and September 2026."),
        ("critical", f"Family plan cancellations rose after its price went from $79 to $89 a week on 1 April 2026. The Family cancellation rate averaged {_pct(pre)} a month from October to March and {_pct(post)} from April to August."),
        ("secondary", f"Support tickets peaked in December 2025 ({int(by_month.support_tickets.max()):,} across all plans) during holiday delivery delays."),
        ("secondary", f"New sign-ups peak in January: {int(by_month.new_signups.max()):,} across all plans in January 2026, the New Year rush."),
        ("critical", f"Family plan active subscribers fell from {a_apr:,} at the start of April to {a_sep:,} at the start of September ({(a_sep / a_apr - 1) * 100:.0f}%). The other three plans all grew over the year."),
        ("secondary", f"Vegetarian boxes have the highest average rating ({rating['Vegetarian']:.2f} out of 5)."),
        ("secondary", f"Single has the lowest cancellation rate, averaging {_pct(churn_by_plan['Single'])} a month."),
        ("caveat", f"September 2026 figures cover 1 to 14 September only, so September sign-ups and cancellations are about half a month. The Family plan's September cancellation rate of {_pct(sep_rate)} is not a recovery."),
        ("secondary", "Figures exclude gift subscriptions, which are about 3% of orders."),
    ]
    quiz = {
        "main": {"gist": "Family-plan cancellations rose (and subscribers fell) after the April price rise",
                 "any": [[r"famil", r"cancel|churn", r"rise|rose|ris|increas|doubl|jump|spik|\bup\b|higher|more|surg|climb|grew"],
                         [r"famil", r"subscri|customer|member", r"fell|fall|drop|declin|\blos|shr|down|fewer"]]},
        "caveat": {"gist": "September covers only 1-14 September",
                   "any": [[r"sep", r"partial|half|incomplete|1\s*(-|–|to)\s*14|first (two|2) weeks|two weeks|not (a )?(full|complete)|mid-|14 days|only"]]},
        "facts": [
            {"q": "What was the Family plan's monthly cancellation rate before the April 2026 price change, and after it? Give both in percent.",
             "fields": [{"name": "before", "answer": round(pre, 2), "tol": 0.6}, {"name": "after", "answer": round(post, 2), "tol": 0.8}]},
            {"q": "How many active Family plan subscribers were there at the start of April 2026, and at the start of September 2026?",
             "fields": [{"name": "april", "answer": int(a_apr), "rel": 0.07}, {"name": "september", "answer": int(a_sep), "rel": 0.07}]}]}
    return _finish("B01", "meal_kit_price_rise", "Meal-kit subscriptions after a price rise",
                   "Harvest Table's pricing committee", "whether to reverse the April price rise on the Family plan",
                   df, columns, findings, quiz)


def brief_b02():
    rng = np.random.default_rng(4402)
    hosp = ["Riverside", "Northgate", "Central", "Eastfield", "Westbrook", "Lakeview", "Bayview", "Hillside"]
    urgent = {"Riverside": 58, "Northgate": 41, "Central": 29, "Eastfield": 27, "Westbrook": 31, "Lakeview": 18, "Bayview": 24, "Hillside": 25}
    visits = {"Riverside": 9800, "Northgate": 8600, "Central": 14800, "Eastfield": 11200, "Westbrook": 6100, "Lakeview": 7300, "Bayview": 9900, "Hillside": 8800}
    lwbs = {"Riverside": 8.9, "Northgate": 3.6, "Central": 3.4, "Eastfield": 3.1, "Westbrook": 3.8, "Lakeview": 2.2, "Bayview": 2.7, "Hillside": 2.9}
    beds = {"Riverside": 34, "Northgate": 30, "Central": 52, "Eastfield": 40, "Westbrook": 18, "Lakeview": 26, "Bayview": 33, "Hillside": 29}
    amb = {"Riverside": 27, "Northgate": 24, "Central": 31, "Eastfield": 38, "Westbrook": 21, "Lakeview": 19, "Bayview": 26, "Hillside": 23}
    cats = {"Urgent": (0.22, 1.0, 0.45, 2.3), "Semi-urgent": (0.45, 2.1, 0.95, 2.35), "Non-urgent": (0.33, 3.1, 1.5, 2.2)}
    rows = []
    for h in hosp:
        for c, (share, wmult, lmult, p90mult) in cats.items():
            med = urgent[h] if c == "Urgent" else round(urgent[h] * wmult + 18 * (c == "Non-urgent") + rng.normal(0, 3))
            rows.append({"hospital": h, "triage_category": c, "visits": int(round(visits[h] * share * rng.uniform(0.96, 1.04))),
                         "median_wait_min": int(med), "p90_wait_min": int(round(med * p90mult * rng.uniform(0.95, 1.05))),
                         "left_without_being_seen_pct": round(lwbs[h] * lmult / 1.0 * rng.uniform(0.95, 1.05), 1),
                         "staffed_beds": beds[h], "ambulance_arrivals_pct": amb[h],
                         "wait_clock_starts_at": "registration" if h == "Northgate" else "triage"})
    df = pd.DataFrame(rows)
    df["lwbs_n"] = df.visits * df.left_without_being_seen_pct / 100
    hl = df.groupby("hospital")[["visits", "lwbs_n"]].sum()
    hl["rate"] = 100 * hl.lwbs_n / hl.visits
    df = df.drop(columns="lwbs_n")
    riv_lwbs, others_lwbs = hl.rate["Riverside"], hl.rate.drop("Riverside").mean()
    others_urgent = np.mean([urgent[h] for h in hosp if h != "Riverside"])
    for c in cats:
        sub = df[df.triage_category == c].set_index("hospital").median_wait_min
        assert sub.idxmin() == "Lakeview" and sub.idxmax() == "Riverside"
    semi = df[df.triage_category == "Semi-urgent"].set_index("hospital")
    long_semi = semi[semi.p90_wait_min > 180].index.tolist()
    non_med = df[df.triage_category == "Non-urgent"].median_wait_min.median()
    assert hl.visits.idxmax() == "Central"
    columns = {
        "hospital": "emergency department", "triage_category": "urgency assigned at triage",
        "visits": "patient visits, July to September 2026", "median_wait_min": "median minutes from arrival to first seen by a doctor",
        "p90_wait_min": "90th percentile of the same wait, minutes", "left_without_being_seen_pct": "percentage of visits that left before being seen",
        "staffed_beds": "staffed emergency beds", "ambulance_arrivals_pct": "percentage of visits arriving by ambulance",
        "wait_clock_starts_at": "event that starts the wait clock"}
    findings = [
        ("secondary", f"Central is the busiest department, with {hl.visits['Central']:,} visits in the quarter."),
        ("secondary", f"Non-urgent patients wait longest everywhere, as expected: a median of {non_med:.0f} minutes across the eight hospitals."),
        ("critical", f"Riverside has the longest median wait for urgent patients: {urgent['Riverside']} minutes, about twice the {others_urgent:.0f}-minute average of the other seven hospitals."),
        ("secondary", f"Eastfield has the highest share of ambulance arrivals ({amb['Eastfield']}%)."),
        ("secondary", f"The slowest 10% of semi-urgent patients waited over three hours at {len(long_semi)} hospitals ({', '.join(long_semi)})."),
        ("caveat", "Northgate starts its wait clock at registration rather than at triage, which adds about 15 minutes. Its 41-minute urgent wait is about 26 minutes on the same basis as the other hospitals, so it is not really the second-slowest."),
        ("secondary", f"Westbrook has the fewest staffed beds ({beds['Westbrook']})."),
        ("critical", f"Riverside also has the most patients leaving without being seen: {riv_lwbs:.1f}% of visits, against an average of {others_lwbs:.1f}% at the other hospitals."),
        ("secondary", "Visits across the network were up 4% on the previous quarter."),
        ("secondary", "Lakeview has the shortest waits in every triage category."),
    ]
    quiz = {
        "main": {"gist": "Riverside has by far the longest urgent waits (and most patients leaving)",
                 "any": [[r"riverside", r"wait|long|slow|delay|worst|left|leav|highest|most"]]},
        "caveat": {"gist": "Northgate measures waits from registration, so it is not comparable",
                   "any": [[r"northgate", r"regist|clock|measur|compar|differen|defin|start|basis|method"]]},
        "facts": [
            {"q": "Which hospital has the longest median wait for urgent patients, and how many minutes is that wait?",
             "fields": [{"name": "hospital", "answer": "Riverside"}, {"name": "minutes", "answer": urgent["Riverside"], "tol": 6}]},
            {"q": "What percentage of Riverside's patients left without being seen, and what is the typical percentage at the other hospitals?",
             "fields": [{"name": "riverside", "answer": round(riv_lwbs, 1), "tol": 1.0}, {"name": "others", "answer": round(others_lwbs, 1), "tol": 1.0}]}]}
    return _finish("B02", "emergency_waits", "Emergency department waits across eight hospitals",
                   "Lakeshore Health's board", "which emergency department should get funding for 12 extra triage nurses",
                   df, columns, findings, quiz)


def brief_b03():
    rng = np.random.default_rng(4403)
    dates = pd.date_range("2026-09-14", "2026-09-27")
    conv = {("control", "desktop"): 3.2, ("control", "mobile"): 2.1, ("control", "tablet"): 2.4,
            ("new_checkout", "desktop"): 3.6, ("new_checkout", "mobile"): 1.5, ("new_checkout", "tablet"): 3.3}
    start = {("control", "desktop"): 6.0, ("control", "mobile"): 5.2, ("control", "tablet"): 5.5,
             ("new_checkout", "desktop"): 6.5, ("new_checkout", "mobile"): 5.4, ("new_checkout", "tablet"): 6.0}
    load = {("control", "desktop"): 2.4, ("control", "mobile"): 2.9, ("control", "tablet"): 2.6,
            ("new_checkout", "desktop"): 1.9, ("new_checkout", "mobile"): 3.0, ("new_checkout", "tablet"): 2.5}
    rows = []
    for d in dates:
        total = rng.normal(30000, 1200) * (1.12 if d.dayofweek >= 5 else 1.0)
        for v in ["control", "new_checkout"]:
            vt = total * (0.501 if v == "control" else 0.499)
            for dev, share in [("desktop", 0.29), ("mobile", 0.68), ("tablet", 0.03)]:
                s = int(round(vt * share * rng.uniform(0.97, 1.03)))
                c = conv[(v, dev)]
                if v == "new_checkout" and dev == "mobile" and d >= pd.Timestamp("2026-09-26"):
                    c = 2.05
                if d.dayofweek == 5:
                    c += 0.3
                if d == pd.Timestamp("2026-09-20"):
                    c *= 0.6
                orders = int(rng.binomial(s, c / 100))
                st = int(rng.binomial(s, start[(v, dev)] / 100))
                rows.append({"date": d.strftime("%Y-%m-%d"), "variant": v, "device": dev, "sessions": s,
                             "checkouts_started": st, "orders": orders, "conversion_pct": round(100 * orders / s, 2),
                             "avg_order_value_usd": round(rng.normal(413, 9), 0), "median_page_load_s": round(load[(v, dev)] + rng.normal(0, 0.05), 2)})
    df = pd.DataFrame(rows)
    g = df.groupby(["variant", "device"])[["sessions", "orders", "checkouts_started"]].sum()
    rate = 100 * g.orders / g.sessions
    tot = df.groupby("variant")[["sessions", "orders", "checkouts_started"]].sum()
    ord_change = (tot.orders["new_checkout"] / tot.orders["control"] - 1) * 100
    mob_share = 100 * df[df.device == "mobile"].sessions.sum() / df.sessions.sum()
    late = df[(df.variant == "new_checkout") & (df.device == "mobile") & (df.date >= "2026-09-26")]
    late_rate = 100 * late.orders.sum() / late.sessions.sum()
    aov = df.groupby("variant").avg_order_value_usd.mean()
    split = 100 * tot.sessions / tot.sessions.sum()
    started_change = (tot.checkouts_started["new_checkout"] / tot.checkouts_started["control"] - 1) * 100
    by_dow = df.assign(dow=pd.to_datetime(df.date).dt.dayofweek).groupby("dow")[["orders", "sessions"]].sum()
    sat = by_dow.orders / by_dow.sessions
    assert sat.idxmax() == 5 and ord_change < -3
    columns = {
        "date": "calendar date", "variant": "control = old checkout, new_checkout = new one-page checkout",
        "device": "visitor's device type", "sessions": "visits to the shop", "checkouts_started": "sessions that opened the checkout",
        "orders": "completed orders", "conversion_pct": "orders as a percentage of sessions",
        "avg_order_value_usd": "average order value, US dollars", "median_page_load_s": "median checkout page load time, seconds"}
    findings = [
        ("secondary", f"Average order value is about the same in both versions ({_money(aov['control'], 0)} old, {_money(aov['new_checkout'], 0)} new)."),
        ("critical", f"On mobile, conversion fell from {rate['control', 'mobile']:.2f}% with the old checkout to {rate['new_checkout', 'mobile']:.2f}% with the new one. On desktop it rose from {rate['control', 'desktop']:.2f}% to {rate['new_checkout', 'desktop']:.2f}%."),
        ("secondary", f"Sessions were split evenly between the versions ({split['control']:.1f}% old, {split['new_checkout']:.1f}% new), as designed."),
        ("secondary", f"Tablet conversion rose from {rate['control', 'tablet']:.2f}% to {rate['new_checkout', 'tablet']:.2f}%, but tablets are only 3% of sessions."),
        ("critical", f"Mobile is {mob_share:.0f}% of sessions, so over the 14 days the new checkout produced {abs(ord_change):.1f}% fewer orders in total ({tot.orders['new_checkout']:,} against {tot.orders['control']:,})."),
        ("secondary", f"More visitors started the new checkout ({started_change:+.1f}% checkouts started)."),
        ("secondary", "The new checkout loads faster on desktop (median 1.9 s against 2.4 s)."),
        ("caveat", f"A bug that hid the payment button on some phones in the new checkout was fixed on 26 September. Mobile conversion on the new checkout was {late_rate:.2f}% on 26 and 27 September, close to the old checkout, but two days is too short to judge."),
        ("secondary", "Conversion is highest on Saturdays in both versions."),
        ("secondary", "Both versions dipped on 20 September during a payment-provider outage."),
    ]
    quiz = {
        "main": {"gist": "The new checkout loses mobile conversions and lowers total orders",
                 "any": [[r"mobile|phone", r"low|fell|fall|drop|declin|hurt|worse|down|decreas|reduc|loss|los"],
                         [r"new (checkout|design|version|variant|flow)|redesign|one-page", r"fewer|lower|fell|drop|declin|worse|down|decreas|reduc|underperform|hurt|los"]]},
        "caveat": {"gist": "A mobile payment-button bug was fixed on 26 September",
                   "any": [[r"bug|fix|button|defect|glitch|error"]]},
        "facts": [
            {"q": "What was the mobile conversion rate with the old checkout and with the new checkout, in percent?",
             "fields": [{"name": "old", "answer": round(rate["control", "mobile"], 2), "tol": 0.15}, {"name": "new", "answer": round(rate["new_checkout", "mobile"], 2), "tol": 0.15}]},
            {"q": "By how many percent did total orders change with the new checkout compared with the old one? Use a negative number for a fall.",
             "fields": [{"name": "change_pct", "answer": round(ord_change, 1), "tol": 2.5}]}]}
    return _finish("B03", "checkout_test", "A/B test of a new checkout page",
                   "Oakline's head of product", "whether to roll out the new one-page checkout to all customers",
                   df, columns, findings, quiz)


def brief_b04():
    rng = np.random.default_rng(4404)
    lines = ["Staff", "Overtime", "Contracts", "Materials", "Fleet"]
    budget = {"Police": [30000, 3500, 4000, 2000, 2500], "Fire": [22000, 2400, 1500, 1200, 1900],
              "Transit": [18000, 2000, 6000, 2500, 4500], "Roads": [9000, 900, 7000, 4000, 2600],
              "Parks": [6000, 300, 3500, 1500, 700], "Library": [5500, 100, 800, 900, 200],
              "Water": [7000, 600, 5000, 3000, 1400], "Housing": [4000, 100, 9000, 400, 100],
              "IT": [6500, 200, 5500, 600, 50], "Admin": [8000, 150, 2200, 300, 100]}
    var = {"Police": [100, 600, -200, 0, 100], "Fire": [200, -50, -100, 150, 100], "Transit": [200, 2400, -100, 200, 400],
           "Roads": [0, 300, 200, 300, 0], "Parks": [-100, 0, -300, 0, 0], "Library": [-150, 0, -50, 0, 0],
           "Water": [100, 100, -100, 100, 0], "Housing": [0, 0, 100, 0, 0], "IT": [0, 0, 350, 0, -50], "Admin": [50, 0, -50, 0, 0]}
    rows = []
    for dept in budget:
        for i, line in enumerate(lines):
            b, v = budget[dept][i], var[dept][i]
            f = b + v
            prior = 2100 if (dept, line) == ("Fire", "Overtime") else round(b * rng.uniform(0.93, 1.0))
            spent = round(f * 7 / 12 * rng.uniform(0.96, 1.04))
            committed = round((f - spent) * (0.6 if line == "Contracts" else 0.1) * rng.uniform(0.8, 1.2))
            rows.append({"department": dept, "budget_line": line, "annual_budget_k": b, "spent_to_date_k": spent,
                         "committed_k": committed, "forecast_year_end_k": f, "variance_k": v,
                         "variance_pct": round(100 * v / b, 1), "prior_year_actual_k": prior})
    df = pd.DataFrame(rows)
    dv = df.groupby("department")[["annual_budget_k", "variance_k", "spent_to_date_k"]].sum()
    total = dv.variance_k.sum() / 1000
    assert abs(total - 4.8) < 1e-9 and dv.annual_budget_k.idxmax() == "Police" and dv.variance_k.idxmin() == "Parks"
    contracts_under = int((df[df.budget_line == "Contracts"].variance_k < 0).sum())
    spent_share = 100 * dv.spent_to_date_k.sum() / dv.annual_budget_k.sum()
    lib_pct = 100 * dv.variance_k["Library"] / dv.annual_budget_k["Library"]
    columns = {
        "department": "city department", "budget_line": "type of spending", "annual_budget_k": "approved budget for the year, $ thousands",
        "spent_to_date_k": "spent from January to July, $ thousands", "committed_k": "ordered but not yet paid, $ thousands",
        "forecast_year_end_k": "forecast spending for the full year, $ thousands", "variance_k": "forecast minus budget, $ thousands (positive = overspend)",
        "variance_pct": "variance as a percentage of budget", "prior_year_actual_k": "actual spending last year, $ thousands"}
    findings = [
        ("secondary", f"Police has the largest budget, {_money(dv.annual_budget_k['Police'] / 1000)} million."),
        ("critical", f"The city forecasts a {_money(total)} million overspend for the year. Transit accounts for {_money(dv.variance_k['Transit'] / 1000)} million of it ({100 * dv.variance_k['Transit'] / 1000 / total:.0f}%); Roads ({_money(dv.variance_k['Roads'] / 1000)} million) and Police ({_money(dv.variance_k['Police'] / 1000)} million) are next."),
        ("secondary", f"Contracts are under budget in {contracts_under} of 10 departments, mostly because invoices arrive late in the year."),
        ("secondary", f"Spending to date is {spent_share:.0f}% of the annual budget after 7 of 12 months (58% would be on track)."),
        ("secondary", "Parks has the largest underspend, $0.4 million."),
        ("caveat", "The Roads forecast assumes snow clearing at the five-year average. A harsh winter could add up to $1.5 million to Roads' overspend."),
        ("secondary", "Fire's overtime forecast is 12% above last year's actual but still within its budget."),
        ("critical", "Most of Transit's overspend is overtime: $2.4 million over a $2.0 million overtime budget (+120%), from driver vacancies covered by extra shifts."),
        ("secondary", "IT's $0.3 million overspend comes from software licence renewals."),
        ("secondary", f"Library is {abs(lib_pct):.1f}% under budget."),
    ]
    quiz = {
        "main": {"gist": "Transit drives most of the city's overspend (mostly overtime)",
                 "any": [[r"transit", r"over|exceed|most|largest|biggest|driv|bulk|majority|account|lead"]]},
        "caveat": {"gist": "A harsh winter could add up to $1.5 million to Roads", "any": [[r"snow|winter|weather"]]},
        "facts": [
            {"q": "What is the city's total forecast overspend, and how much of it comes from Transit? Give both in $ millions.",
             "fields": [{"name": "total", "answer": 4.8, "tol": 0.4}, {"name": "transit", "answer": 3.1, "tol": 0.3}]},
            {"q": "How much of Transit's overspend comes from overtime, in $ millions?",
             "fields": [{"name": "overtime", "answer": 2.4, "tol": 0.3}]}]}
    return _finish("B04", "city_overspend", "A city's mid-year budget forecast",
                   "the City of Marlow's finance committee", "which department to put under a mid-year spending review",
                   df, columns, findings, quiz)


def brief_b05():
    rng = np.random.default_rng(4405)
    spec = {"Swiftline": (90, 15.2, 13, 9.10), "Northway": (90, 20.5, 2, 9.55), "ParcelPro": (40, 17.4, 1, 9.30)}
    rows, k = [], 0
    for courier, (n, base, n_tail, cost) in spec.items():
        tail_idx = set(rng.choice(n, n_tail, replace=False).tolist())
        first = pd.Timestamp("2026-09-01") if courier == "ParcelPro" else pd.Timestamp("2026-08-03")
        days = (pd.Timestamp("2026-10-30") - first).days
        for i in range(n):
            k += 1
            region = rng.choice(["urban", "suburban", "rural"], p=[0.5, 0.3, 0.2])
            dist = {"urban": rng.uniform(3, 25), "suburban": rng.uniform(15, 60), "rural": rng.uniform(50, 150)}[region]
            ship = first + pd.Timedelta(days=int(rng.integers(0, days + 1)))
            hours = base * np.exp(rng.normal(0, 0.18)) + 0.045 * dist + (3.0 if ship.dayofweek == 0 else 0.0)
            hours = min(hours, 46.5)
            if i in tail_idx:
                hours = rng.uniform(49, 76)
            promised = 36 if region == "rural" else 24
            rows.append({"delivery_id": f"D{k:04d}", "courier": courier, "ship_date": ship.strftime("%Y-%m-%d"),
                         "weekday": ship.day_name(), "region": region, "distance_km": round(dist, 1),
                         "promised_hours": promised, "actual_hours": round(hours, 1), "late": bool(hours > promised),
                         "damaged": False, "cost_usd": round(cost + 0.004 * dist + rng.normal(0, 0.15), 2),
                         "customer_rating": int(np.clip(round(rng.normal(4.4, 0.6)), 1, 5))})
    df = pd.DataFrame(rows).sort_values(["ship_date", "delivery_id"]).reset_index(drop=True)
    for c in spec:  # one damaged parcel per courier
        df.loc[df[df.courier == c].index[3], "damaged"] = True
    g = df.groupby("courier")
    over = 100 * g.actual_hours.apply(lambda s: (s > 48).mean())
    med = g.actual_hours.median()
    assert med.idxmin() == "Swiftline" and over.idxmax() == "Swiftline"
    cost = g.cost_usd.mean()
    rating = g.customer_rating.mean()
    rural_gap = df[df.region == "rural"].actual_hours.median() - df[df.region != "rural"].actual_hours.median()
    wd = df.groupby("weekday").actual_hours.median()
    late = 100 * df.late.mean()
    r = df[df.actual_hours < 48][["distance_km", "actual_hours"]].corr().iloc[0, 1]
    columns = {
        "delivery_id": "delivery reference", "courier": "courier company", "ship_date": "date the parcel left the pharmacy",
        "weekday": "weekday of ship_date", "region": "delivery area type", "distance_km": "road distance, km",
        "promised_hours": "delivery time promised to the customer, hours", "actual_hours": "hours from dispatch to delivery",
        "late": "delivered after the promised time", "damaged": "parcel arrived damaged", "cost_usd": "price paid to the courier, US dollars",
        "customer_rating": "customer rating of the delivery, 1 to 5"}
    findings = [
        ("secondary", f"Courier prices are within 50 cents of each other ({_money(cost.min(), 2)} to {_money(cost.max(), 2)} per delivery)."),
        ("secondary", f"Customer ratings are similar ({rating.min():.1f} to {rating.max():.1f} out of 5)."),
        ("critical", f"Cold packs keep these medicines safe for 48 hours. Swiftline delivered {over['Swiftline']:.1f}% of parcels after more than 48 hours, against {over['Northway']:.1f}% for Northway and {over['ParcelPro']:.1f}% for ParcelPro."),
        ("secondary", "Each courier damaged one parcel."),
        ("secondary", f"Rural deliveries take about {rural_gap:.0f} hours longer than urban and suburban ones."),
        ("critical", f"Swiftline is fastest on a typical delivery: a median of {med['Swiftline']:.1f} hours, against {med['Northway']:.1f} for Northway and {med['ParcelPro']:.1f} for ParcelPro."),
        ("secondary", "Parcels shipped on Mondays take longest, because of the weekend backlog."),
        ("secondary", f"Overall, {late:.0f}% of deliveries arrived after the time promised to the customer."),
        ("caveat", "ParcelPro only started in September, so its figures rest on 40 deliveries, against 90 for each of the others. Its single delivery over 48 hours could easily be two or three in a larger sample."),
        ("secondary", f"Delivery time rises with distance (correlation {r:.2f} for deliveries under 48 hours)."),
    ]
    quiz = {
        "main": {"gist": "Swiftline is fast on a typical day but too often breaks the 48-hour limit",
                 "any": [[r"swiftline", r"48|late|slow|tail|long|unreliab|variab|risk|delay|exceed|over|spoil|outlier|inconsist|spread"],
                         [r"northway|parcelpro|parcel pro", r"reliab|safe|best|fewer|consistent|within|predictab"]]},
        "caveat": {"gist": "ParcelPro's numbers rest on only 40 deliveries",
                   "any": [[r"parcelpro|parcel pro", r"\b40\b|small|few|limited|sample|only|started|new|short|less data"]]},
        "facts": [
            {"q": "What percentage of each courier's deliveries took more than 48 hours?",
             "fields": [{"name": "swiftline", "answer": round(over["Swiftline"], 1), "tol": 3.0},
                        {"name": "northway", "answer": round(over["Northway"], 1), "tol": 2.0},
                        {"name": "parcelpro", "answer": round(over["ParcelPro"], 1), "tol": 2.5}]},
            {"q": "What is the median delivery time, in hours, for Swiftline and for Northway?",
             "fields": [{"name": "swiftline", "answer": round(med["Swiftline"], 1), "tol": 2.5}, {"name": "northway", "answer": round(med["Northway"], 1), "tol": 2.5}]}]}
    return _finish("B05", "courier_delivery_times", "Delivery times of three couriers for refrigerated medicine",
                   "Medley Pharmacy's operations director", "which courier to contract for refrigerated medicine deliveries next year",
                   df, columns, findings, quiz)


def brief_b06():
    rng = np.random.default_rng(4406)
    towns = ["Ashby", "Brockton", "Carlow", "Denham", "Elstree", "Fenwick", "Garston", "Harlow", "Ingleby", "Jarrow",
             "Kelso", "Langley", "Morley", "Norton", "Oakham", "Pelham", "Radley", "Selby", "Thorne", "Upton", "Wexley", "Yarrow"]
    rows = []
    for i in range(46):
        large = i >= 35
        new_large = large and i >= 42
        sq = int(round(rng.uniform(38000, 55000) if large else rng.uniform(18000, 32000), -2))
        opened = int(rng.integers(2025, 2027) if new_large else (rng.integers(2012, 2022) if large else rng.integers(1996, 2023)))
        if new_large:
            spsf = rng.normal(390, 20)
        elif large:
            spsf = rng.normal(500, 22)
        else:
            spsf = rng.normal(585, 35) + (45 if opened < 2010 else 0)
        if i == 38:
            sq, spsf = 54800, 535.0          # the chain's top seller is a large store
        sales = sq * spsf / 1e6
        rows.append({"store_id": f"S{i + 1:02d}", "town": towns[i % len(towns)], "format": "large" if large else "standard",
                     "sq_ft": sq, "opened_year": opened, "annual_sales_musd": round(sales, 2), "sales_per_sqft_usd": round(spsf),
                     "staff": int(round(sales * 6.2 * rng.uniform(0.92, 1.08))), "parking_spaces": int(round(sq / 1000 * 9 * rng.uniform(0.85, 1.15))),
                     "catchment_pop_k": int(round(rng.uniform(85, 140) if large else rng.uniform(30, 90)))})
    df = pd.DataFrame(rows).sample(frac=1, random_state=7).reset_index(drop=True)
    g = df.groupby("format")
    spsf_m, sales_m, sq_m = g.sales_per_sqft_usd.mean(), g.annual_sales_musd.mean(), g.sq_ft.mean()
    est = df[(df.format == "large") & (df.opened_year < 2025)].sales_per_sqft_usd.mean()
    assert est < spsf_m["standard"] and (df.format.eq("large") & (df.opened_year >= 2025)).sum() == 4
    top = df.loc[df.annual_sales_musd.idxmax()]
    assert top.format == "large"
    std = df[df.format == "standard"]
    old_gap = std[std.opened_year < 2010].sales_per_sqft_usd.mean() - std[std.opened_year >= 2010].sales_per_sqft_usd.mean()
    r_park = df[["sq_ft", "parking_spaces"]].corr().iloc[0, 1]
    columns = {
        "store_id": "store code", "town": "town", "format": "standard (under 33,000 sq ft) or large (over 38,000 sq ft)",
        "sq_ft": "selling floor area, square feet", "opened_year": "year the store opened",
        "annual_sales_musd": "sales in the last 12 months, $ millions", "sales_per_sqft_usd": "annual sales per square foot, US dollars",
        "staff": "employees", "parking_spaces": "car park spaces", "catchment_pop_k": "people living within 15 minutes' drive, thousands"}
    findings = [
        ("secondary", "Large stores sit in bigger catchments: 85,000 to 140,000 people within 15 minutes, against 30,000 to 90,000 for standard stores."),
        ("critical", f"Large stores sell less per square foot: {_money(spsf_m['large'], 0)} a year on average, against {_money(spsf_m['standard'], 0)} for standard stores."),
        ("secondary", f"Parking spaces rise with floor area (correlation {r_park:.2f})."),
        ("secondary", f"The chain's top seller is {top.store_id} in {top.town}, a large store with {_money(top.annual_sales_musd)} million in sales."),
        ("critical", f"A large store averages {_money(sales_m['large'])} million in annual sales against {_money(sales_m['standard'])} million for a standard store: {100 * (sales_m['large'] / sales_m['standard'] - 1):.0f}% more sales for {100 * (sq_m['large'] / sq_m['standard'] - 1):.0f}% more floor space."),
        ("secondary", f"Standard stores opened before 2010 sell about {_money(old_gap, 0)} more per square foot than newer standard stores."),
        ("secondary", "Staff numbers track sales closely, at about 6 employees per $1 million of sales in both formats."),
        ("caveat", f"4 of the 11 large stores opened in 2025 or 2026 and are still building up trade. Without them, large stores average {_money(est, 0)} per square foot, still below standard stores."),
        ("secondary", f"The average store has {df.sq_ft.mean():,.0f} square feet of selling space."),
        ("secondary", "Two stores closed for six weeks of renovation; their sales are annualised."),
    ]
    quiz = {
        "main": {"gist": "Bigger stores sell less per square foot; large format is a poor use of space",
                 "any": [[r"large|big|size", r"less|lower|fewer|declin|fall|drop|worse|diminish|smaller|under|weaker|not|reduc"],
                         [r"standard|small", r"more|higher|better|outperform|stronger|efficien"]]},
        "caveat": {"gist": "Four large stores are new and still ramping up",
                   "any": [[r"new|ramp|recent|opened|young|matur|building up|2025|2026"]]},
        "facts": [
            {"q": "What are average annual sales per square foot for standard stores and for large stores, in US dollars?",
             "fields": [{"name": "standard", "answer": round(spsf_m["standard"]), "tol": 30}, {"name": "large", "answer": round(spsf_m["large"]), "tol": 30}]},
            {"q": "What are average annual sales per store, in $ millions, for standard stores and for large stores?",
             "fields": [{"name": "standard", "answer": round(sales_m["standard"], 1), "tol": 1.2}, {"name": "large", "answer": round(sales_m["large"], 1), "tol": 1.5}]}]}
    return _finish("B06", "store_size", "Sales per square foot by store size",
                   "FreshWay's property committee", "whether the next five stores should be large-format (over 38,000 sq ft) or standard size",
                   df, columns, findings, quiz)


def brief_b07():
    rng = np.random.default_rng(4407)
    quarters = ["2025Q1", "2025Q2", "2025Q3", "2025Q4", "2026Q1", "2026Q2", "2026Q3"]
    cats = {  # base return rate, units per quarter, price, margin, usual reason
        "Footwear": (11.0, 9000, 74, 48, "Size/fit"), "Swimwear": (24.0, 5200, 38, 55, "Size/fit"),
        "Dresses": (18.0, 7800, 59, 52, "Style"), "Denim": (9.0, 8800, 64, 62, "Size/fit"),
        "Tops": (8.0, 21000, 26, 54, "Style"), "Knitwear": (7.0, 7400, 52, 50, "Quality"),
        "Outerwear": (10.0, 4200, 129, 46, "Size/fit"), "Activewear": (12.0, 9600, 44, 51, "Size/fit"),
        "Lingerie": (6.0, 6800, 31, 58, "Size/fit"), "Accessories": (3.0, 12500, 22, 60, "Changed mind"),
        "Kids": (9.0, 5600, 24, 47, "Size/fit"), "Sleepwear": (5.0, 4300, 33, 53, "Style")}
    rows = []
    for cat, (rate, units, price, margin, reason) in cats.items():
        for qi, q in enumerate(quarters):
            if cat == "Kids" and q == "2025Q1":
                continue
            r = rate + rng.normal(0, 0.35)
            defect = rng.uniform(5, 12)
            top_reason = reason
            if cat == "Dresses":
                r = 18.0 - 2.0 * max(0, qi - 2) / 4 + rng.normal(0, 0.2)
            if cat == "Footwear":
                defect = rng.uniform(7.5, 10.5)
                if q == "2026Q3":
                    r, defect, top_reason = 19.2, 46.0, "Defect"
            u = units * (1.8 if (cat == "Outerwear" and q.endswith("Q4")) else 1.0) * rng.uniform(0.94, 1.06)
            u = int(round(u, -1))
            ret = int(round(u * r / 100))
            rows.append({"category": cat, "quarter": q, "units_sold": u, "units_returned": ret,
                         "return_rate_pct": round(100 * ret / u, 1), "defect_share_of_returns_pct": round(defect, 1),
                         "top_return_reason": top_reason, "avg_price_usd": price, "gross_margin_pct": margin})
    df = pd.DataFrame(rows)
    fw = df[df.category == "Footwear"].set_index("quarter")
    before = fw.loc[quarters[:-1], "return_rate_pct"].mean()
    q3 = fw.loc["2026Q3", "return_rate_pct"]
    defect_before = fw.loc[quarters[:-1], "defect_share_of_returns_pct"].mean()
    piv = df.pivot(index="category", columns="quarter", values="return_rate_pct")
    other_moves = (piv["2026Q3"] - piv[["2026Q1", "2026Q2"]].mean(axis=1)).drop("Footwear").abs().max()
    assert (piv.idxmax() == "Swimwear").all() and (piv.idxmin() == "Accessories").all()
    by_q = df.groupby("quarter")[["units_returned", "units_sold"]].sum()
    overall = 100 * by_q.units_returned / by_q.units_sold
    dresses = piv.loc["Dresses"]
    columns = {
        "category": "product category", "quarter": "calendar quarter", "units_sold": "units sold", "units_returned": "units returned",
        "return_rate_pct": "units returned as a percentage of units sold", "defect_share_of_returns_pct": "percentage of returns where the customer reported a defect",
        "top_return_reason": "most common return reason", "avg_price_usd": "average selling price, US dollars", "gross_margin_pct": "gross margin, percent"}
    findings = [
        ("secondary", f"Swimwear has the highest return rate every quarter, around {piv.loc['Swimwear'].mean():.0f}%, mostly for fit."),
        ("secondary", f"Accessories have the lowest return rate, around {piv.loc['Accessories'].mean():.0f}%."),
        ("critical", f"Footwear's return rate jumped to {q3:.1f}% in Q3 2026, from an average of {before:.1f}% over the previous six quarters. No other category moved by more than {other_moves:.1f} points."),
        ("secondary", "Outerwear sales peak every fourth quarter."),
        ("secondary", f"The overall return rate is steady, between {overall.min():.1f}% and {overall.max():.1f}% each quarter."),
        ("critical", f"Defects drove the footwear jump: {fw.loc['2026Q3', 'defect_share_of_returns_pct']:.0f}% of footwear returns in Q3 2026 reported a defect (mostly soles coming apart), against an average of {defect_before:.0f}% before."),
        ("secondary", "Denim has the highest gross margin (62%)."),
        ("caveat", "Footwear's return window was extended from 30 to 60 days for a Q3 2026 promotion. Past extensions added about 2 points to the return rate, so the policy explains only a small part of the jump."),
        ("secondary", f"Dresses' return rate has fallen by about 2 points over the last year, to {dresses['2026Q3']:.1f}%."),
        ("secondary", "Kids clothing launched in Q2 2025, so it has one quarter less history."),
    ]
    quiz = {
        "main": {"gist": "Footwear returns jumped in Q3 2026, driven by defects",
                 "any": [[r"footwear|shoe", r"return|defect|jump|spike|rise|rose|increas|doubl|surg"]]},
        "caveat": {"gist": "Footwear's return window was extended to 60 days in Q3",
                   "any": [[r"window|60|policy|promotion|extended|longer"]]},
        "facts": [
            {"q": "What was footwear's return rate in Q3 2026, and what was it typically in the quarters before? Give both in percent.",
             "fields": [{"name": "q3_2026", "answer": round(q3, 1), "tol": 1.5}, {"name": "before", "answer": round(before, 1), "tol": 1.5}]},
            {"q": "What percentage of footwear returns reported a defect in Q3 2026?",
             "fields": [{"name": "defect_share", "answer": round(fw.loc["2026Q3", "defect_share_of_returns_pct"], 1), "tol": 5}]}]}
    return _finish("B07", "returns_jump", "Product return rates by category",
                   "Thread & Co's merchandising director", "which product category to investigate first for a quality problem",
                   df, columns, findings, quiz)


def brief_b08():
    rng = np.random.default_rng(4408)
    days = pd.date_range("2025-12-01", "2026-02-28")
    snap = {"2026-01-12": (-24, 928), "2026-01-13": (-26, 951), "2026-01-14": (-29, 1012), "2026-01-15": (-25, 935)}
    cold_extra = {"2025-12-22": -17, "2026-02-03": -18, "2026-02-04": -16}
    rows = []
    for d in days:
        ds = d.strftime("%Y-%m-%d")
        season = -8 if d.month == 1 else -4
        tmin = float(np.clip(round(rng.normal(season, 4.5)), -14.5, 6))
        tmin = snap[ds][0] if ds in snap else cold_extra.get(ds, tmin)
        wk = 0.92 if d.dayofweek >= 5 else 1.0
        core = 640 + 9.0 * max(0, 2 - tmin)
        forecast = core * wk * (1.015 if tmin > -3 else 1.0)
        boost = 0.0055 * max(0, -12 - tmin)
        actual = core * wk * (1 + boost) + rng.normal(0, 7)
        if ds in snap:
            actual = snap[ds][1]
        cap = 910 if pd.Timestamp("2026-01-10") <= d <= pd.Timestamp("2026-01-16") else 960
        rows.append({"date": ds, "weekday": d.day_name(), "min_temp_c": tmin, "avg_temp_c": round(tmin + 5 + rng.normal(0, 1), 1),
                     "wind_kmh": int(rng.integers(5, 46)), "forecast_peak_mw": int(round(forecast)), "actual_peak_mw": int(round(actual)),
                     "available_capacity_mw": cap, "imports_mw": int(round(max(0, actual - cap) + rng.uniform(5, 35))),
                     "residential_share_pct": round(52 + 0.4 * max(0, -tmin) + rng.normal(0, 0.8), 1)})
    df = pd.DataFrame(rows)
    err = 100 * (df.actual_peak_mw / df.forecast_peak_mw - 1)
    cold = df.min_temp_c < -15
    cold_pct, cold_mw = err[cold].mean(), (df.actual_peak_mw - df.forecast_peak_mw)[cold].mean()
    mild = err[df.min_temp_c > -3].mean()
    exceed = df[df.actual_peak_mw > df.available_capacity_mw]
    assert exceed.date.tolist() == list(snap) and (df.actual_peak_mw > 960).sum() == 1
    mape = err.abs().mean()
    wk = df.assign(we=pd.to_datetime(df.date).dt.dayofweek >= 5).groupby("we").actual_peak_mw.mean()
    r_wind = df[["wind_kmh", "actual_peak_mw"]].corr().iloc[0, 1]
    columns = {
        "date": "calendar date", "weekday": "day of the week", "min_temp_c": "lowest temperature of the day, °C", "avg_temp_c": "average temperature, °C",
        "wind_kmh": "average wind speed, km/h", "forecast_peak_mw": "day-ahead forecast of peak demand, MW", "actual_peak_mw": "actual peak demand, MW",
        "available_capacity_mw": "generating capacity available that day, MW", "imports_mw": "power bought from neighbouring grids at the peak, MW",
        "residential_share_pct": "share of peak demand from homes, percent"}
    findings = [
        ("secondary", f"Across all 90 days the forecast was off by {mape:.1f}% on average."),
        ("secondary", f"Wind speed has almost no relation to peak demand (correlation {r_wind:.2f})."),
        ("critical", f"The forecast under-predicts demand on very cold days. On the {int(cold.sum())} days below −15 °C, actual peak demand was {cold_pct:.1f}% above the forecast on average (about {cold_mw:.0f} MW)."),
        ("secondary", f"Weekend peaks are about {100 * (1 - wk[True] / wk[False]):.0f}% lower than weekday peaks."),
        ("secondary", "The share of demand from homes rises in cold weather."),
        ("critical", "Peak demand exceeded available capacity on 4 days, 12 to 15 January, and the shortfall was covered with emergency imports."),
        ("secondary", f"Imports at the daily peak averaged {df.imports_mw.mean():.0f} MW."),
        ("caveat", "Unit 3 (50 MW) was out for repair from 10 to 16 January, cutting available capacity from 960 to 910 MW. With Unit 3 running, only 14 January would have exceeded capacity."),
        ("secondary", f"The highest peak of the winter was {df.actual_peak_mw.max():,} MW on 14 January."),
        ("secondary", f"On mild days (above −3 °C) the forecast slightly over-predicts, by {abs(mild):.1f}% on average."),
    ]
    quiz = {
        "main": {"gist": "Demand outruns the forecast on very cold days and exceeded capacity in the January cold snap",
                 "any": [[r"cold|freez|low temp|below|frigid|snap|winter", r"under|exceed|above|higher|more than|short|miss|surpass|outstrip|beyond|outpac"],
                         [r"capacity", r"exceed|short|above|over|surpass|beyond|insufficient|outstrip"]]},
        "caveat": {"gist": "Unit 3 was out for repair, so capacity was 50 MW lower that week",
                   "any": [[r"unit\s*3|unit three|generator|repair|maintenance|outage|offline|out of service"]]},
        "facts": [
            {"q": "On the coldest days, by roughly what percentage did actual peak demand exceed the forecast?",
             "fields": [{"name": "pct", "answer": round(cold_pct, 1), "tol": 2.5}]},
            {"q": "On how many days did peak demand exceed the available capacity?",
             "fields": [{"name": "days", "answer": 4, "tol": 0}]}]}
    return _finish("B08", "winter_peak_demand", "Winter electricity demand against the forecast",
                   "Brightwater Electric's planning team", "whether to buy 40 MW of extra peak capacity for this winter",
                   df, columns, findings, quiz)


def brief_b09():
    rng = np.random.default_rng(4409)
    depts = {"Customer Support": 140, "Engineering": 210, "Sales": 95, "Marketing": 45, "Finance": 22,
             "Product": 60, "Operations": 80, "People": 25, "Design": 30}
    items = {"recommend": "I would recommend Northlight as a place to work", "manager": "My manager supports my growth",
             "tools": "I have the tools I need to do my job", "workload": "My workload is manageable",
             "two_years": "I see myself working here in two years", "strategy": "I understand the company's strategy"}
    rows = []
    for d, hc in depts.items():
        rr25 = rng.uniform(75, 90) if d != "Customer Support" else 81.0
        rr26 = rng.uniform(74, 90) if d != "Customer Support" else 38.0
        for key, text in items.items():
            v25 = rng.uniform(60, 80)
            ch = rng.uniform(-4, 3)
            if key == "strategy":
                v25, ch = rng.uniform(52, 62), rng.uniform(7, 11)
            if key == "tools" and d == "Engineering":
                v25, ch = 86.0, 2.0
            if key == "workload" and d == "Sales":
                v25, ch = 39.0, 0.0
            if key == "workload" and d != "Sales":
                v25 = rng.uniform(52, 70)
            if d == "Marketing" and key == "recommend":
                v25, ch = 66.0, 8.0
            if d == "Customer Support":
                v25, ch = {"manager": (71, -27), "two_years": (64, -23), "recommend": (72, -12)}.get(key, (v25, ch))
            for year, val, rr in [(2025, v25, rr25), (2026, v25 + ch, rr26)]:
                rows.append({"department": d, "question": text, "year": year, "pct_favourable": round(val),
                             "n_responses": int(round(hc * rr / 100)), "headcount": hc, "response_rate_pct": round(rr)})
    df = pd.DataFrame(rows)
    piv = df.pivot_table(index=["department", "question"], columns="year", values="pct_favourable")
    change = piv[2026] - piv[2025]
    assert change.idxmin() == ("Customer Support", items["manager"])
    assert change.drop(("Customer Support", items["manager"])).min() > -27
    by_y = df.assign(w=df.pct_favourable * df.n_responses).groupby("year")[["w", "n_responses"]].sum()
    comp = by_y.w / by_y.n_responses
    strat = change.xs(items["strategy"], level="question").mean()
    work = df[(df.question == items["workload"]) & (df.year == 2026)].set_index("department").pct_favourable
    assert work.idxmin() == "Sales"
    s = piv.loc["Customer Support"]
    columns = {
        "department": "department", "question": "survey statement", "year": "survey year (September)",
        "pct_favourable": "percentage who agreed or strongly agreed", "n_responses": "people who answered",
        "headcount": "people in the department", "response_rate_pct": "n_responses as a percentage of headcount"}
    findings = [
        ("secondary", f"Company-wide, {comp[2026]:.0f}% of answers were favourable this year, against {comp[2025]:.0f}% last year."),
        ("critical", f"Customer Support's score on 'My manager supports my growth' fell from {s.loc[items['manager'], 2025]:.0f}% to {s.loc[items['manager'], 2026]:.0f}% favourable, a {s.loc[items['manager'], 2025] - s.loc[items['manager'], 2026]:.0f}-point drop and the largest change for any department on any question."),
        ("secondary", "Engineering scores highest on 'I have the tools I need' (88%)."),
        ("secondary", f"Sales has the lowest score on 'My workload is manageable' ({work['Sales']:.0f}%), unchanged from last year."),
        ("critical", f"Support's score on 'I see myself working here in two years' fell from {s.loc[items['two_years'], 2025]:.0f}% to {s.loc[items['two_years'], 2026]:.0f}%, a warning sign for staff turnover."),
        ("secondary", f"Understanding of company strategy rose about {strat:.0f} points across departments after this year's town halls."),
        ("secondary", "Finance is the smallest department, with 22 people."),
        ("caveat", "Only 38% of Support staff answered this year, against 81% last year, so Support's results are less certain than other departments'."),
        ("secondary", "Marketing improved most on 'I would recommend Northlight', up 8 points."),
        ("secondary", "The survey ran from 8 to 22 September."),
    ]
    quiz = {
        "main": {"gist": "Customer Support's scores collapsed, especially manager support and intent to stay",
                 "any": [[r"support", r"manager|growth|drop|fell|declin|lower|stay|leav|retention|worse|plung|fall|collaps|sharp"]]},
        "caveat": {"gist": "Support's response rate fell from 81% to 38%",
                   "any": [[r"respon|answered|fewer|\b38|sample|particip"]]},
        "facts": [
            {"q": "What percentage of Customer Support staff answered favourably to 'My manager supports my growth' in 2025 and in 2026?",
             "fields": [{"name": "y2025", "answer": float(s.loc[items["manager"], 2025]), "tol": 4}, {"name": "y2026", "answer": float(s.loc[items["manager"], 2026]), "tol": 4}]},
            {"q": "What percentage of Customer Support staff answered favourably to 'I see myself working here in two years' in 2025 and in 2026?",
             "fields": [{"name": "y2025", "answer": float(s.loc[items["two_years"], 2025]), "tol": 4}, {"name": "y2026", "answer": float(s.loc[items["two_years"], 2026]), "tol": 4}]}]}
    return _finish("B09", "staff_survey", "An annual staff survey across nine departments",
                   "Northlight Software's leadership team", "which department gets a manager-support programme this quarter",
                   df, columns, findings, quiz)


def brief_b10():
    rng = np.random.default_rng(4410)
    sizes = {"2026-01": 18400, "2026-02": 9800, "2026-03": 11200, "2026-04": 10900, "2026-05": 10400,
             "2026-06": 15600, "2026-07": 16100, "2026-08": 14900, "2026-09": 11300}
    old = {1: 70, 7: 48, 14: 34, 30: 22, 60: 16, 90: 13}
    new = {1: 70, 7: 52, 14: 41, 30: 31, 60: 24, 90: 20}
    sep = {1: 70, 7: 51, 14: 39, 30: 28.5}
    pull = pd.Timestamp("2026-11-02")
    rows = []
    for m, n in sizes.items():
        version = "new" if m >= "2026-06" else "old"
        curve = sep if m == "2026-09" else (new if version == "new" else old)
        paid = (3.7 if m == "2026-09" else 3.9 if version == "new" else 3.1) + rng.normal(0, 0.08)
        lessons = (15.9 if version == "new" else 14.2) * rng.uniform(0.97, 1.03)
        end = pd.Period(m, "M").end_time.normalize()
        for day, pct in curve.items():
            if end + pd.Timedelta(days=day) > pull:
                continue
            p = pct + (rng.normal(0, 0.6) if day > 1 else rng.normal(0, 0.8))
            rows.append({"signup_month": m, "onboarding_version": version, "cohort_size": n, "day": day,
                         "retained_users": int(round(n * p / 100)), "retention_pct": round(p, 1),
                         "paid_conversion_30d_pct": round(paid, 2), "avg_lessons_first_30d": round(lessons, 1)})
    df = pd.DataFrame(rows)
    d30 = df[df.day == 30].set_index("signup_month")
    old30 = d30.loc[[m for m in sizes if m < "2026-06"], "retention_pct"].mean()
    new30 = d30.loc[["2026-06", "2026-07", "2026-08"], "retention_pct"].mean()
    coh = df.drop_duplicates("signup_month").set_index("signup_month")
    paid_old = coh.loc[[m for m in sizes if m < "2026-06"], "paid_conversion_30d_pct"].mean()
    paid_new = coh.loc[["2026-06", "2026-07", "2026-08"], "paid_conversion_30d_pct"].mean()
    les = coh.groupby("onboarding_version").avg_lessons_first_30d.mean()
    d90 = df[(df.day == 90) & (df.onboarding_version == "old")].retention_pct.mean()
    columns = {
        "signup_month": "month the users signed up (cohort)", "onboarding_version": "onboarding flow the cohort saw",
        "cohort_size": "users who signed up that month", "day": "days since sign-up", "retained_users": "users still active on that day",
        "retention_pct": "retained_users as a percentage of cohort_size", "paid_conversion_30d_pct": "percentage of the cohort who bought a subscription within 30 days",
        "avg_lessons_first_30d": "average lessons completed per user in the first 30 days"}
    findings = [
        ("secondary", "The January cohort is the largest (18,400 sign-ups), the New Year rush."),
        ("critical", f"Day-30 retention rose from an average of {old30:.1f}% for the January to May cohorts (old onboarding) to {new30:.1f}% for June to August (new onboarding)."),
        ("secondary", "Day-1 retention is about 70% for every cohort, before and after the change."),
        ("secondary", f"Lessons completed in the first 30 days rose {100 * (les['new'] / les['old'] - 1):.0f}% with the new onboarding."),
        ("secondary", f"Only about {d90:.0f}% of early-2026 users are still active after 90 days."),
        ("caveat", f"A TV campaign ran from June to August, the same months as the new onboarding, so part of the gain may come from different kinds of users. The September cohort, after the campaign, still reached {d30.loc['2026-09', 'retention_pct']:.1f}% at day 30."),
        ("secondary", "February had the smallest cohort (9,800)."),
        ("critical", f"Paid conversion within 30 days rose from {paid_old:.1f}% to {paid_new:.1f}% over the same cohorts."),
        ("secondary", "Data runs to 2 November 2026, so the September cohort is only measured to day 30 and August to day 60."),
        ("secondary", "A user counts as retained on day N if they completed a lesson on day N or later."),
    ]
    quiz = {
        "main": {"gist": "The new onboarding raised day-30 retention (and paid conversion)",
                 "any": [[r"onboard|new (flow|version)|june|redesign|change", r"retention|retain|stay|kept|return|stick|churn|active", r"\bup\b|rose|higher|improv|increas|better|boost|gain|lift|jump"]]},
        "caveat": {"gist": "A TV campaign ran at the same time as the new onboarding",
                   "any": [[r"campaign|\btv\b|advert|marketing|television"]]},
        "facts": [
            {"q": "What was day-30 retention for cohorts with the old onboarding and with the new onboarding, in percent?",
             "fields": [{"name": "old", "answer": round(old30, 1), "tol": 2.0}, {"name": "new", "answer": round(new30, 1), "tol": 2.0}]},
            {"q": "What was the 30-day paid conversion rate before and after the new onboarding, in percent?",
             "fields": [{"name": "old", "answer": round(paid_old, 2), "tol": 0.3}, {"name": "new", "answer": round(paid_new, 2), "tol": 0.3}]}]}
    return _finish("B10", "onboarding_retention", "App retention before and after a new onboarding flow",
                   "Lingo Loop's growth team", "whether to keep the new onboarding flow launched on 1 June or go back to the old one",
                   df, columns, findings, quiz)


def brief_b11():
    rng = np.random.default_rng(4411)
    months = [f"2026-{m:02d}" for m in range(1, 10)]
    spec = {  # applications(month index), approval %, avg loan $k, acquisition cost $, days to fund, default %, basis
        "Branch": (lambda i: 900, 66, 20, 820, 6, 1.8, "actual"),
        "Online": (lambda i: 1200 + 52.5 * i, 60, 15, 340, 9, 2.4, "actual"),
        "Broker": (lambda i: 520, 41, 58, 1450, 14, 2.1, "projected"),
        "Partner app": (lambda i: 300 + 50 * (i - 2), 72, 9, 260, 8, 3.0, "projected")}
    rows = []
    for i, m in enumerate(months):
        for ch, (apps_f, appr, loan, acq, days, dflt, basis) in spec.items():
            if ch == "Partner app" and i < 2:
                continue
            apps = int(round(apps_f(i) * (1.2 if m == "2026-03" else 1.0) * rng.uniform(0.96, 1.04)))
            approved = int(round(apps * (appr + rng.normal(0, 1.2)) / 100))
            funded = int(round(approved * 0.93))
            avg = loan * rng.uniform(0.97, 1.03)
            rows.append({"month": m, "channel": ch, "applications": apps, "approved": approved, "funded_loans": funded,
                         "funded_value_musd": round(funded * avg / 1000, 2), "avg_loan_k": round(avg, 1),
                         "acquisition_cost_per_loan_usd": int(round(acq * rng.uniform(0.97, 1.03))), "days_to_fund": days,
                         "default_rate_12m_pct": dflt, "default_rate_basis": basis})
    df = pd.DataFrame(rows)
    g = df.groupby("channel")[["applications", "funded_value_musd", "funded_loans", "approved"]].sum()
    vshare = 100 * g.funded_value_musd / g.funded_value_musd.sum()
    ashare = 100 * g.applications / g.applications.sum()
    acq = df.groupby("channel").acquisition_cost_per_loan_usd.mean()
    on = df[df.channel == "Online"].set_index("month").applications
    total_apps = df.groupby("month").applications.sum()
    assert total_apps.idxmax() == "2026-03"
    columns = {
        "month": "month", "channel": "where the application came from", "applications": "loan applications received",
        "approved": "applications approved", "funded_loans": "loans paid out", "funded_value_musd": "value of loans paid out, $ millions",
        "avg_loan_k": "average loan paid out, $ thousands", "acquisition_cost_per_loan_usd": "marketing, commission and processing cost per funded loan, US dollars",
        "days_to_fund": "median days from application to payout", "default_rate_12m_pct": "share of loans in default within 12 months, percent",
        "default_rate_basis": "actual = observed history, projected = model estimate"}
    findings = [
        ("secondary", f"Online applications grew {100 * (on.iloc[-1] / on.iloc[0] - 1):.0f}% from January to September."),
        ("secondary", "Branches pay out fastest, in a median of 6 days."),
        ("critical", f"Broker brings {vshare['Broker']:.0f}% of funded loan value from {ashare['Broker']:.0f}% of applications, because broker loans are much larger ($58,000 on average, against $9,000 to $20,000 in the other channels)."),
        ("secondary", "Partner-app loans are the smallest, about $9,000 on average."),
        ("secondary", "Approval rates are steady from month to month in every channel."),
        ("caveat", "Broker loans only started in November 2025, so the broker 12-month default rate (2.1%) is a projection, not an observed figure. The same is true of the partner app."),
        ("secondary", "Applications peaked in March, when tax refunds arrive."),
        ("critical", f"Broker is the most expensive channel: about {_money(acq['Broker'], 0)} to acquire each funded loan, against {_money(acq['Branch'], 0)} at branches, {_money(acq['Online'], 0)} online and {_money(acq['Partner app'], 0)} through the partner app."),
        ("secondary", "The partner app launched in March 2026."),
        ("secondary", f"Total funded value over the nine months was {_money(g.funded_value_musd.sum(), 0)} million."),
    ]
    quiz = {
        "main": {"gist": "Broker is costly per loan but brings a large share of loan value",
                 "any": [[r"broker", r"value|large|bigger|dollar|share|most|volume|third|account|revenue|significant|outsized|dispropor|big"]]},
        "caveat": {"gist": "Broker default rates are projections, not observed",
                   "any": [[r"default", r"project|estimat|not yet|too new|unknown|early|forecast|expected|assum|model"]]},
        "facts": [
            {"q": "What percentage of funded loan value, and what percentage of applications, came from the broker channel?",
             "fields": [{"name": "value_share", "answer": round(vshare["Broker"], 1), "tol": 4}, {"name": "application_share", "answer": round(ashare["Broker"], 1), "tol": 3}]},
            {"q": "What does it cost to acquire one funded loan through the broker channel and through branches, in US dollars?",
             "fields": [{"name": "broker", "answer": round(acq["Broker"]), "tol": 150}, {"name": "branch", "answer": round(acq["Branch"]), "tol": 100}]}]}
    return _finish("B11", "loan_channels", "Loan applications and costs by channel",
                   "Ridgeview Credit Union's lending committee", "whether to close the broker channel to cut acquisition costs",
                   df, columns, findings, quiz)


def brief_b12():
    rng = np.random.default_rng(4412)
    months = ["2026-04", "2026-05", "2026-06", "2026-07", "2026-08", "2026-09"]
    days = {"2026-04": 30, "2026-05": 31, "2026-06": 30, "2026-07": 31, "2026-08": 31, "2026-09": 30}
    season = {"2026-04": 0.97, "2026-05": 1.03, "2026-06": 1.0, "2026-07": 1.02, "2026-08": 1.04, "2026-09": 0.98}
    temp = {"2026-04": 12, "2026-05": 16, "2026-06": 20, "2026-07": 23, "2026-08": 25, "2026-09": 19}
    rain = {"2026-04": 61, "2026-05": 54, "2026-06": 41, "2026-07": 18, "2026-08": 31, "2026-09": 46}
    notices = {"2026-06": 600, "2026-07": 1240, "2026-08": 900, "2026-09": 450}
    dist = {  # households, baseline L/household/day, cut after 1 June, smart meter %
        "Hillcrest": (6800, 428, 0.095, 48), "Lakeside": (14200, 340, 0.22, 66), "Orchard": (5200, 262, 0.18, 71),
        "Millbrook": (9100, 355, 0.24, 58), "Stonebridge": (7400, 390, 0.27, 55), "Riverbend": (10300, 310, 0.20, 62),
        "Elmwood": (8600, 372, 0.25, 57), "Fairview": (9800, 335, 0.21, 63)}
    hh_total = sum(v[0] for v in dist.values())
    rows = []
    for dname, (hh, base, cut, smart) in dist.items():
        for m in months:
            restricted = m >= "2026-06"
            lphd = base * season[m] * ((1 - cut) if restricted else 1.0)
            if dname == "Hillcrest" and m >= "2026-07":
                lphd *= 1.08
            lphd *= rng.uniform(0.99, 1.01)
            rows.append({"district": dname, "month": m, "households": hh, "total_use_ml": round(hh * lphd * days[m] / 1e6, 1),
                         "litres_per_household_day": int(round(lphd)), "target_lphd": int(round(0.8 * base)),
                         "rainfall_mm": int(round(rain[m] * rng.uniform(0.9, 1.1))), "avg_temp_c": temp[m],
                         "enforcement_notices": int(round(notices.get(m, 0) * hh / hh_total * rng.uniform(0.8, 1.2))),
                         "smart_meter_pct": smart})
    df = pd.DataFrame(rows)
    pre = df[df.month < "2026-06"].groupby("district").litres_per_household_day.mean()
    post = df[df.month >= "2026-06"].groupby("district").litres_per_household_day.mean()
    cut = 100 * (1 - post / pre)
    others = cut.drop("Hillcrest")
    assert cut.idxmin() == "Hillcrest" and others.min() > 16 and others.max() < 28
    hc_true = 100 * (1 - (post["Hillcrest"] - 0.75 * post["Hillcrest"] * (1 - 1 / 1.08)) / pre["Hillcrest"])
    daily = df.assign(per_day=df.total_use_ml / df.month.map(days), after=df.month >= "2026-06")
    tot = daily.groupby("after").per_day.sum() / daily.groupby("after").month.nunique()
    smart_all = (df.drop_duplicates("district").smart_meter_pct * df.drop_duplicates("district").households).sum() / hh_total
    columns = {
        "district": "water district", "month": "month", "households": "households served", "total_use_ml": "water delivered, megalitres",
        "litres_per_household_day": "average litres per household per day", "target_lphd": "district target under restrictions, litres per household per day (20% below April-May)",
        "rainfall_mm": "rainfall, mm", "avg_temp_c": "average temperature, °C", "enforcement_notices": "warning notices issued for breaking restrictions",
        "smart_meter_pct": "percentage of homes with smart meters"}
    findings = [
        ("secondary", "July rainfall was about half the normal amount."),
        ("critical", f"Seven of the eight districts cut use per household by {others.min():.0f}% to {others.max():.0f}% after restrictions began on 1 June. Hillcrest cut by only {cut['Hillcrest']:.0f}%."),
        ("secondary", "Lakeside serves the most households (14,200)."),
        ("secondary", "Enforcement notices peaked in July, at about 1,240 across the valley."),
        ("secondary", f"About {smart_all:.0f}% of homes have smart meters."),
        ("critical", f"Hillcrest households used {post['Hillcrest']:.0f} litres a day from June to September, against a target of {int(round(0.8 * 428))} litres (20% below their April-May level)."),
        ("secondary", "August was the hottest month, averaging 25 °C."),
        ("caveat", f"Hillcrest's water meters were replaced in July, and the new meters read about 8% higher than the old ones. Allowing for that, Hillcrest's cut is about {hc_true:.0f}%, still well short of the 20% target."),
        ("secondary", f"Total water use across the valley fell {100 * (1 - tot[True] / tot[False]):.0f}% after restrictions began."),
        ("secondary", f"Orchard has the lowest use per household ({post['Orchard']:.0f} litres a day under restrictions)."),
    ]
    quiz = {
        "main": {"gist": "Hillcrest barely cut its water use while the other districts cut about a fifth",
                 "any": [[r"hillcrest", r"\bnot\b|barely|only|little|miss|short|fail|lag|above|over|least|small|behind|minimal|far|outlier|exceed|lowest|smallest|didn"]]},
        "caveat": {"gist": "Hillcrest's new meters read about 8% higher", "any": [[r"meter|reading"]]},
        "facts": [
            {"q": "By what percentage did Hillcrest cut water use per household after the restrictions, and roughly what was the cut in the other districts?",
             "fields": [{"name": "hillcrest", "answer": round(cut["Hillcrest"], 1), "tol": 3}, {"name": "others", "answer": round(others.mean(), 1), "tol": 3}]},
            {"q": "How many litres per household per day did Hillcrest use after the restrictions began, and what was its target?",
             "fields": [{"name": "after", "answer": round(post["Hillcrest"]), "tol": 20}, {"name": "target", "answer": int(round(0.8 * 428)), "tol": 20}]}]}
    return _finish("B12", "water_restrictions", "Water use by district under summer restrictions",
                   "the Clearwater Valley water board", "where to focus enforcement if the summer watering restrictions are extended",
                   df, columns, findings, quiz)


BRIEF_FUNCS = [brief_b01, brief_b02, brief_b03, brief_b04, brief_b05, brief_b06,
               brief_b07, brief_b08, brief_b09, brief_b10, brief_b11, brief_b12]


def freeze_briefs():
    """Write each brief's data.csv and brief.json; on a replay, check the frozen files match the generator."""
    briefs = []
    for fn in BRIEF_FUNCS[:N_BRIEFS]:
        b = fn()
        folder = BRIEF_DIR / b["id"]
        folder.mkdir(parents=True, exist_ok=True)
        csv_text = b["df"].to_csv(index=False, lineterminator="\n")
        meta = {k: v for k, v in b.items() if k != "df"}
        meta["n_rows"] = len(b["df"])
        meta["csv_sha256"] = hashlib.sha256(csv_text.encode("utf-8")).hexdigest()
        meta_text = json.dumps(meta, indent=1, ensure_ascii=False, default=float) + "\n"
        for name, text in [("data.csv", csv_text), ("brief.json", meta_text)]:
            path = folder / name
            if path.exists() and path.read_text(encoding="utf-8") != text:
                raise RuntimeError(f"{path} differs from the generator; the frozen brief would change")
            if not path.exists():
                path.write_text(text, encoding="utf-8", newline="\n")
        briefs.append(b)
    return briefs


BRIEFS = freeze_briefs()
BRIEF = {b["id"]: b for b in BRIEFS}
summary = pd.DataFrame([{"brief": b["id"], "topic": b["topic"], "rows": len(b["df"]), "columns": b["df"].shape[1],
                         "decision": b["decision"]} for b in BRIEFS])
display(summary)
````

## 3. Five levels of instruction

Every prompt has three parts. The brief (data, column descriptions and notes) and a fixed output contract are the same at every level. The contract asks for matplotlib code that reads `data.csv` and draws one figure of 10 by 6 inches, which may contain several panels; the harness saves the figure. Only the task part changes:

| Level | Task text | What it adds |
|---|---|---|
| L0, bare request | "Make a chart from this." | Nothing. The model must infer the purpose. |
| L1, purpose | Who the audience is, what they must decide, and that they will see the chart for about ten seconds | The decision that defines what is critical |
| L2, design principles | L1 plus six general rules: a title that states the finding, show only what the message needs, one strong colour and grey for context, number labels only where the number is the message, state a caveat that changes the reading, honest scales | General advice that names no note |
| L3, priorities | L2 plus the three critical notes quoted, and "the other notes are true but not needed; leave them off" | The editing decision, made for the model |
| L4, look and revise | The L2 prompt, the model's own L2 code, and a picture of the chart it produced: "Look at it as the audience would see it, then improve the chart" | A second pass with sight of the output |

The comparison between L2 and L3 separates two skills. At L2 the model has to decide what matters; at L3 it is told. If L3 charts are much cleaner and more complete than L2 charts, the gap is the model's editorial judgement. If L3 charts are still cluttered, the problem is in how the model draws, not in what it chooses. L4 starts from the same L2 prompt, so the difference between L2 and L4 is the effect of one look at the rendered chart.

The cell below prints the L0 prompt for B01 (with the data shortened) and the L3 task text.

````python
# Every prompt has the same three parts: the brief (data and analyst notes), the task (which grows with
# the instruction level) and a fixed output contract. Only the task part changes between L0 and L3.
# L4 reuses the L2 prompt and adds the model's own L2 code and a picture of the chart it produced.

SYSTEM_PROMPT = "You are a helpful assistant."

CONTRACT = """## Output
Write Python code that draws the chart with matplotlib. pandas and numpy are available; no other libraries.
- Read the data from the file data.csv in the working directory.
- Make one figure with figsize=(10, 6). It may contain more than one panel.
- Do not call plt.show() or savefig(); the figure is saved for you.
Reply with the code in a single ```python code block."""

PRINCIPLES = """Design guidance:
- Pick the one message the audience most needs for this decision, and make the title state it as a sentence.
- Show only the data needed for that message. Leave out series, columns and notes that do not help the decision.
- Use one strong colour for what matters and grey for context. Label lines or bars directly instead of using a legend when you can.
- Add a number label only where the number is part of the message. Do not label every bar or point.
- If a caveat changes how the chart should be read, state it briefly on the chart.
- Keep scales honest: bar charts start at zero, units are labelled, no 3-D effects and no second y-axis."""


def brief_block(b):
    csv_text = (BRIEF_DIR / b["id"] / "data.csv").read_text(encoding="utf-8").strip()
    cols = "\n".join(f"- {c}: {d}" for c, d in b["columns"].items())
    notes = "\n".join(f"- {f['text']}" for f in b["findings"])
    return (f"## Data: data.csv ({len(b['df'])} rows)\nColumns:\n{cols}\n\n```csv\n{csv_text}\n```\n\n"
            f"## Analyst notes\n{notes}")


def task_block(b, level):
    purpose = (f"Make a chart for {b['audience']}, who must decide {b['decision']}. They will see it for about "
               f"ten seconds on a slide in a meeting, so it should help them make that decision.")
    if level == "L0":
        return "## Task\nMake a chart from this."
    if level == "L1":
        return f"## Task\n{purpose}"
    if level in ("L2", "L4"):
        return f"## Task\n{purpose}\n\n{PRINCIPLES}"
    if level == "L3":
        need = "\n".join(f"- {f['text']}" for f in b["findings"] if f["role"] != "secondary")
        return (f"## Task\n{purpose}\n\n{PRINCIPLES}\n\nPriorities for this decision. The audience needs these points:\n{need}\n"
                "The other analyst notes are true but not needed for this decision. Leave them off the chart "
                "unless the chart cannot be read correctly without them.")
    raise ValueError(level)


def draft_prompt(b, level):
    base_level = "L2" if level == "L4" else level
    return f"{brief_block(b)}\n\n{task_block(b, base_level)}\n\n{CONTRACT}"


def revise_prompt(b, first_code):
    return (f"{draft_prompt(b, 'L2')}\n\n## Your first version\n```python\n{first_code}\n```\n"
            "The attached image is the chart this code produced. Look at it as the audience would see it, then improve "
            "the chart. Reply with the complete revised code in a single ```python code block.")


def repair_prompt(original_prompt, code, error):
    return (f"{original_prompt}\n\n## Your code\n```python\n{code}\n```\nRunning it failed:\n```\n{error}\n```\n"
            "Fix the code. Reply with the complete corrected code in a single ```python code block.")


def extract_code(text):
    """The longest fenced python block, or the whole reply if it is bare code. None means unreadable."""
    blocks = re.findall(r"```(?:python|py|Python)?[ \t]*\r?\n(.*?)```", text or "", flags=re.S)
    blocks = [b for b in blocks if "plt" in b or "matplotlib" in b]
    if blocks:
        return max(blocks, key=len).strip() + "\n"
    if text and "import matplotlib" in text and "```" not in text:
        return text.strip() + "\n"
    return None


print(f"Prompt sizes for {BRIEFS[0]['id']} (characters):",
      {lv: len(draft_prompt(BRIEFS[0], lv)) for lv in ["L0", "L1", "L2", "L3"]})
````

## 5. Making the charts

For each brief, model and level, the harness sends one prompt and takes the longest Python code block from the reply. It runs the code in a separate Python process inside a folder that holds only `data.csv`, saves the figure as a PNG and counts what is on it (section 7). If the code raises an error or draws no figure, the model gets one repair call with its code and the error, which is what a person would do next. A reply with no code in it is counted as unreadable and not repaired.

The L4 call goes to the same model after its L2 chart has rendered. It contains the L2 prompt, the model's L2 code and the L2 chart as an image.

The three cells below define the call layer (cache, command-line wrappers, tracing), the renderer and the generation loop.

````python
# Model calls. Every reply is cached in data/cache/calls.jsonl, keyed by a hash of the full request
# (model, settings, system prompt, prompt text and the SHA-256 of every attached image). A replay reads
# the cache; a cache miss with LIVE = False is an error, so a replay can never spend money.

NOTEBOOK = "44-chart-overload"
LANGFUSE_HOST = os.environ.get("LANGFUSE_HOST", "http://localhost:3300").rstrip("/")
LANGFUSE_AUTH = (os.environ.get("LANGFUSE_PUBLIC_KEY", ""), os.environ.get("LANGFUSE_SECRET_KEY", ""))


def _langfuse_running():
    if not (LIVE and all(LANGFUSE_AUTH)):
        return False
    try:
        return requests.get(f"{LANGFUSE_HOST}/api/public/health", timeout=2).ok
    except requests.RequestException:
        return False


LANGFUSE_ON = _langfuse_running()


def trace_call(cache_key, name, model, request, response, start, end, usage=None, cost_usd=None,
               session_id=None, metadata=None, error=None):
    """Send one fresh model call to the local Langfuse (if it is running). Never raises."""
    if not LANGFUSE_ON:
        return

    def attr(k, v):
        return {"key": k, "value": {"stringValue": v if isinstance(v, str) else json.dumps(v, default=str)}}

    tags = {"key": "langfuse.trace.tags", "value": {"arrayValue": {"values": [{"stringValue": f"nb{NOTEBOOK[:2]}"}]}}}
    attrs = [attr("langfuse.observation.type", "generation"), attr("langfuse.observation.model.name", model),
             attr("langfuse.observation.input", request), attr("langfuse.observation.output", response),
             attr("langfuse.trace.name", f"{NOTEBOOK[:2]} {name}"), tags,
             attr("langfuse.environment", "notebook"), attr("langfuse.trace.metadata.notebook", NOTEBOOK),
             attr("langfuse.trace.metadata.cache_key", cache_key)]
    if usage:
        attrs.append(attr("langfuse.observation.usage_details", usage))
    if cost_usd is not None:
        attrs.append(attr("langfuse.observation.cost_details", {"total": cost_usd}))
    if session_id:
        attrs.append(attr("langfuse.session.id", f"{NOTEBOOK}/{session_id}"))
    if error:
        attrs += [attr("langfuse.observation.level", "ERROR"), attr("langfuse.observation.status_message", str(error))]
    for k, v in (metadata or {}).items():
        attrs.append(attr(f"langfuse.observation.metadata.{k}", v))
    span = {"traceId": hashlib.sha256(f"{NOTEBOOK}|{cache_key}".encode()).hexdigest()[:32],
            "spanId": hashlib.sha256(f"{NOTEBOOK}|{cache_key}|gen|0".encode()).hexdigest()[:16],
            "name": name, "kind": 1, "attributes": attrs,
            "startTimeUnixNano": str(int(start.timestamp() * 1e9)), "endTimeUnixNano": str(int(end.timestamp() * 1e9))}
    body = {"resourceSpans": [{"resource": {"attributes": [attr("service.name", "experiment-notebooks")]},
                               "scopeSpans": [{"scope": {"name": NOTEBOOK}, "spans": [span]}]}]}
    try:
        requests.post(f"{LANGFUSE_HOST}/api/public/otel/v1/traces", json=body, auth=LANGFUSE_AUTH, timeout=10)
    except requests.RequestException:
        pass


def _clean_env():
    env = {k: v for k, v in os.environ.items() if k not in ("CLAUDECODE", "FORCE_COLOR")}
    env["NO_COLOR"] = "1"
    return env


def _run(cmd, inp, cwd, timeout):
    try:
        p = subprocess.run(cmd, input=inp, text=True, encoding="utf-8", errors="replace", capture_output=True,
                           timeout=timeout, cwd=cwd, env=_clean_env())
        return p.stdout, p.stderr, p.returncode, None
    except subprocess.TimeoutExpired as e:
        dec = lambda x: x.decode(errors="replace") if isinstance(x, bytes) else (x or "")
        return dec(e.stdout), dec(e.stderr), None, "timeout"


def call_cli(model, prompt, images, timeout=900):
    """One call through claude -p, codex exec or muse exec, in an empty folder with no tools."""
    backend = model["backend"]
    out = {"text": "", "usage": None, "cost_usd": None}
    with tempfile.TemporaryDirectory(prefix="nb44_") as tmp:
        if backend == "claude":
            cmd = [shutil.which("claude"), "-p", "--model", model["id"], "--tools", "", "--safe-mode",
                   "--no-session-persistence", "--system-prompt", SYSTEM_PROMPT, "--effort", model["effort"],
                   "--input-format", "stream-json", "--output-format", "stream-json", "--verbose", "--max-budget-usd", "3"]
            content = [{"type": "image", "source": {"type": "base64", "media_type": "image/png",
                                                    "data": base64.b64encode(Path(p).read_bytes()).decode()}} for p in images]
            content.append({"type": "text", "text": prompt})
            inp = json.dumps({"type": "user", "message": {"role": "user", "content": content}}) + "\n"
        elif backend == "codex":
            cmd = [shutil.which("codex.cmd") or shutil.which("codex"), "exec", "--skip-git-repo-check", "--ephemeral",
                   "--ignore-user-config", "--sandbox", "read-only", "-m", model["id"],
                   "-c", f'model_reasoning_effort="{model["effort"]}"', "-c", 'web_search="disabled"', "--json"]
            for feature in ["shell_tool", "unified_exec", "apps", "plugins", "multi_agent", "browser_use",
                            "computer_use", "image_generation", "code_mode", "code_mode_host", "shell_snapshot"]:
                cmd += ["--disable", feature]
            for p in images:
                cmd += ["-i", str(Path(p).resolve())]
            cmd += ["-"]
            inp = prompt
        elif backend == "muse":
            pf = Path(tmp) / "prompt.txt"
            pf.write_text(prompt, encoding="utf-8")
            cmd = [shutil.which("muse.cmd") or shutil.which("muse"), "exec", "--preset", "empty", "--no-session-log",
                   "--disable-web-tools", "--disable-reminders", "--disable-request-user-input",
                   "--no-foreign-personal-context", "--disable-write", "--disable-shell",
                   "--reasoning-effort", model["effort"], "--max-model-steps", "1", "--model", model["id"],
                   "--prompt-file", str(pf)]
            for p in images:
                cmd += ["--image", str(Path(p).resolve())]
            inp = None
        else:
            raise ValueError(backend)
        raw, stderr, code, err = _run(cmd, inp, tmp, timeout)
    out.update(raw=raw[-200000:], stderr=(stderr or "")[-4000:], returncode=code)
    if err:
        out["error"] = err
        return out
    if backend == "claude":
        events = [json.loads(l) for l in raw.splitlines() if l.startswith("{")]
        finals = [e for e in events if e.get("type") == "result"]
        if finals:
            f = finals[-1]
            out.update(text=f.get("result", "") or "", usage=f.get("usage"), cost_usd=f.get("total_cost_usd"))
            if f.get("is_error"):
                out["error"] = f.get("result") or "claude error"
        out["raw"] = ""          # the stream repeats the reply; keep the cache small
    elif backend == "codex":
        events = [json.loads(l) for l in raw.splitlines() if l.startswith("{")]
        msgs = [e["item"]["text"] for e in events if e.get("type") == "item.completed" and e["item"].get("type") == "agent_message"]
        out["text"] = msgs[-1] if msgs else ""
        out["usage"] = next((e["usage"] for e in events if e.get("type") == "turn.completed"), None)
        fails = [e for e in events if e.get("type") in ("turn.failed", "error")]
        if fails:
            out["error"] = json.dumps(fails[-1])[:500]
        out["raw"] = ""
    else:
        out["text"] = raw.strip()
        out["raw"] = ""
    if not out.get("error") and (code or not out["text"]):
        out["error"] = f"CLI exit {code} or empty reply"
    return out


def call_openrouter(model, prompt, images, timeout=900):
    content = [{"type": "text", "text": prompt}]
    for p in images:
        content.append({"type": "image_url", "image_url": {"url": "data:image/png;base64," + base64.b64encode(Path(p).read_bytes()).decode()}})
    body = {"model": model["id"], "messages": [{"role": "system", "content": SYSTEM_PROMPT}, {"role": "user", "content": content}],
            "provider": {"order": [model["provider"]], "allow_fallbacks": False}, "max_tokens": 32000, "usage": {"include": True}}
    if model["effort"] != "default":
        body["reasoning"] = {"effort": model["effort"]}
    out = {"text": "", "usage": None, "cost_usd": None}
    for attempt in range(6):
        try:
            r = requests.post("https://openrouter.ai/api/v1/chat/completions", json=body, timeout=timeout,
                              headers={"Authorization": "Bearer " + os.environ["OPENROUTER_API_KEY"]})
        except requests.RequestException as e:
            out["error"] = f"request failed: {e}"
            time.sleep(10 * (attempt + 1))
            continue
        if r.status_code in (429, 500, 502, 503, 504):
            out["error"] = f"HTTP {r.status_code}: {r.text[:300]}"
            time.sleep(15 * (attempt + 1))
            continue
        d = r.json()
        if "choices" not in d:
            out["error"] = f"HTTP {r.status_code}: {json.dumps(d)[:500]}"
            return out
        msg = d["choices"][0]["message"]
        out = {"text": msg.get("content") or "", "usage": d.get("usage"), "cost_usd": (d.get("usage") or {}).get("cost"),
               "provider": d.get("provider"), "served_model": d.get("model"), "finish_reason": d["choices"][0].get("finish_reason")}
        if not out["text"]:
            out["error"] = f"empty reply (finish_reason {out['finish_reason']})"
        return out
    return out


class CallCache:
    def __init__(self, path):
        self.path = Path(path)
        self.rows, self.lock, self.key_locks, self.fresh = {}, threading.Lock(), {}, 0
        if self.path.exists():
            for line in self.path.read_text(encoding="utf-8").splitlines():
                row = json.loads(line)
                assert digest(row["request"]) == row["key"], "corrupt cache row"
                self.rows[row["key"]] = row          # a later row for the same key (a retried error) wins

    def ask(self, model, prompt, images=(), name="call", session=None):
        images = [Path(p) for p in images]
        request = {"notebook": NOTEBOOK, "model": {k: model[k] for k in ("backend", "id", "effort") if k in model},
                   "provider": model.get("provider"), "system": SYSTEM_PROMPT, "prompt": prompt,
                   "images": [{"path": p.relative_to(ROOT).as_posix(), "sha256": sha_file(p)} for p in images]}
        key = digest(request)
        with self.lock:
            klock = self.key_locks.setdefault(key, threading.Lock())
        with klock:
            row = self.rows.get(key)
            if row and not row["response"].get("error"):
                return row
            if not LIVE:
                if row:
                    return row
                raise RuntimeError(f"cache miss for {name} (set NB44_LIVE=1 to call models)")
            with self.lock:
                if self.fresh >= MAX_NEW_CALLS:
                    raise RuntimeError("MAX_NEW_CALLS reached")
                self.fresh += 1
            start, failed = now(), []
            for attempt in range(3):          # transient failures (empty Muse step, timeout, HTTP 5xx) get two retries
                if model["backend"] == "openrouter":
                    resp = call_openrouter(model, prompt, images)
                else:
                    resp = call_cli(model, prompt, images)
                if not resp.get("error"):
                    break
                if "usage limit" in str(resp["error"]).lower():
                    raise RuntimeError(f"{model['id']} usage limit reached: {str(resp['error'])[:200]}")
                failed.append(str(resp["error"])[:300])
                time.sleep(5 * (attempt + 1))
            resp["failed_attempts"] = failed
            end = now()
            resp["elapsed_s"] = round((end - start).total_seconds(), 1)
            row = {"key": key, "name": name, "request": request, "response": resp, "collected_at": end.isoformat()}
            with self.lock:
                with self.path.open("a", encoding="utf-8", newline="\n") as f:
                    f.write(json.dumps(row, ensure_ascii=False) + "\n")
                self.rows[key] = row
            usage = resp.get("usage") or {}
            trace_call(key, name, model["id"], {"prompt": prompt, "images": request["images"], "system": SYSTEM_PROMPT},
                       resp.get("text", ""), start, end, cost_usd=resp.get("cost_usd"), session_id=session,
                       metadata={"backend": model["backend"]}, error=resp.get("error"))
            print(f"{end:%H:%M:%S} {name:<34} {model['id']:<34} {'ERROR ' + str(resp.get('error'))[:80] if resp.get('error') else 'ok'} {resp['elapsed_s']}s", flush=True)
            return row


CACHE = CallCache(CACHE_DIR / "calls.jsonl")
LIMITS = {"claude": threading.Semaphore(4), "muse": threading.Semaphore(4), "openrouter": threading.Semaphore(6),
          "codex": threading.Semaphore(2)}


def ask(model, prompt, images=(), name="call", session=None):
    with LIMITS[model["backend"]]:
        return CACHE.ask(model, prompt, images, name, session)


print("Cached replies:", len(CACHE.rows), "| Langfuse tracing:", LANGFUSE_ON)
````

````python
# Rendering. The model's code runs in a separate Python process inside a temporary folder that holds only
# data.csv. After the code finishes, the runner saves the figure and walks its matplotlib objects to count
# what the reader will see: every drawn text item by role, numbers printed on the chart, data series,
# colours, panels and overlapping labels.

RUNNER = r'''
import json, re, sys, traceback, warnings
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.figure as mfig
from matplotlib.text import Text
from matplotlib.lines import Line2D
from matplotlib.collections import PathCollection, PolyCollection, LineCollection, QuadMesh
from matplotlib.container import BarContainer
from matplotlib.patches import Wedge, Rectangle
import matplotlib.colors as mcolors

_save = mfig.Figure.savefig
mfig.Figure.savefig = lambda self, *a, **k: None
plt.savefig = lambda *a, **k: None
plt.show = lambda *a, **k: None
plt.close = lambda *a, **k: None
out = {"ok": False}
try:
    src = open("chart_code.py", encoding="utf-8").read()
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        exec(compile(src, "chart_code.py", "exec"), {"__name__": "__main__"})
except SystemExit as e:
    if e.code not in (0, None):
        out["error"] = f"SystemExit({e.code})"
except BaseException as e:
    tb = traceback.format_exception(type(e), e, e.__traceback__)
    out["error"] = "".join(tb[-4:])[-1500:]
if "error" not in out:
    nums = plt.get_fignums()
    if not nums:
        out["error"] = "The code ran but created no matplotlib figure."
    else:
        figs = [plt.figure(n) for n in nums]
        fig = max(figs, key=lambda f: len(f.findobj()))
        out["n_figures"] = len(figs)
        _save(fig, "chart.png", dpi=100, bbox_inches="tight", pad_inches=0.15, facecolor="white",
              metadata={"Software": None})
        r = fig.canvas.get_renderer()
        W, H = fig.bbox.width, fig.bbox.height

        def visible(a):
            while a is not None:
                if not a.get_visible():
                    return False
                a = getattr(a, "axes", None) if a is not getattr(a, "axes", None) else None
            return True

        role, drawn_ticks = {}, set()
        for ax in fig.axes:
            for t in (ax.title, getattr(ax, "_left_title", None), getattr(ax, "_right_title", None)):
                if t is not None:
                    role[id(t)] = "title"
            for axis in (ax.xaxis, ax.yaxis):
                role[id(axis.label)] = "axis_label"
                role[id(axis.get_offset_text())] = "tick"
                for tk in axis.get_major_ticks() + axis.get_minor_ticks():
                    role[id(tk.label1)] = role[id(tk.label2)] = "tick"
                if ax.axison and axis.get_visible():
                    for tk in axis._update_ticks():
                        for lab in (tk.label1, tk.label2):
                            if lab.get_visible() and lab.get_text().strip():
                                drawn_ticks.add(id(lab))
            leg = ax.get_legend()
            if leg is not None:
                for t in leg.get_texts() + [leg.get_title()]:
                    role[id(t)] = "legend"
        for leg in fig.legends:
            for t in leg.get_texts() + [leg.get_title()]:
                role[id(t)] = "legend"
        for name in ("_suptitle", "_supxlabel", "_supylabel"):
            t = getattr(fig, name, None)
            if t is not None:
                role[id(t)] = "title" if name == "_suptitle" else "axis_label"

        texts, seen = [], set()
        for t in fig.findobj(Text):
            if id(t) in seen:
                continue
            seen.add(id(t))
            s = t.get_text().strip()
            rl = role.get(id(t), "annotation")
            if not s or not t.get_visible() or (t.get_alpha() is not None and t.get_alpha() == 0):
                continue
            if rl == "tick" and id(t) not in drawn_ticks:
                continue
            if t.axes is not None and not t.axes.get_visible():
                continue
            if rl == "legend":
                leg = t.axes.get_legend() if t.axes is not None else None
                if leg is not None and not leg.get_visible():
                    continue
            try:
                bb = t.get_window_extent(r)
            except Exception:
                continue
            if bb.width <= 0 or bb.height <= 0:
                continue
            texts.append({"role": rl, "text": s[:200], "size": float(t.get_fontsize()),
                          "bbox": [float(bb.x0), float(bb.y0), float(bb.x1), float(bb.y1)]})

        def overlap(a, b):
            w = min(a[2], b[2]) - max(a[0], b[0]); h = min(a[3], b[3]) - max(a[1], b[1])
            if w <= 0 or h <= 0:
                return False
            small = min((a[2] - a[0]) * (a[3] - a[1]), (b[2] - b[0]) * (b[3] - b[1]))
            return w * h > 0.15 * small
        n_overlap = sum(overlap(texts[i]["bbox"], texts[j]["bbox"]) for i in range(len(texts)) for j in range(i + 1, len(texts)))

        def hexes(c):
            try:
                arr = mcolors.to_rgba_array(c)
            except Exception:
                return []
            res = []
            for rgba in arr:
                if rgba[3] < 0.05 or min(rgba[:3]) > 0.94:
                    continue
                res.append(mcolors.to_hex(rgba[:3]))
            return res

        panels, series, bars, ref_lines, colors = 0, 0, 0, 0, set()
        for ax in fig.axes:
            if not ax.get_visible() or ax.get_label() == "<colorbar>":
                continue
            has = False
            for ln in ax.lines:
                if not ln.get_visible():
                    continue
                n = len(ln.get_xdata()) if hasattr(ln.get_xdata(), "__len__") else 1
                if n > 2:
                    series += 1; has = True; colors.update(hexes(ln.get_color()))
                else:
                    ref_lines += 1
            for cont in ax.containers:
                if isinstance(cont, BarContainer):
                    vis = [p for p in cont.patches if p.get_visible()]
                    if vis:
                        series += 1; bars += len(vis); has = True
                        for p in vis:
                            colors.update(hexes(p.get_facecolor()))
            for col in ax.collections:
                if not col.get_visible():
                    continue
                if isinstance(col, PathCollection) and len(col.get_offsets()) > 0:
                    series += 1; has = True; colors.update(hexes(col.get_facecolor()))
                elif isinstance(col, (PolyCollection, QuadMesh)):
                    has = True; colors.update(hexes(col.get_facecolor())[:3])
                elif isinstance(col, LineCollection):
                    ref_lines += 1
            for p in ax.patches:
                if isinstance(p, Wedge) and p.get_visible():
                    has = True; colors.update(hexes(p.get_facecolor()))
            if ax.images:
                has = True
            panels += int(has)

        def is_grey(h):
            rgb = mcolors.to_rgb(h)
            return max(rgb) - min(rgb) < 0.08
        roles = [t["role"] for t in texts]
        ann = [t for t in texts if t["role"] == "annotation"]
        out.update(ok=True, figsize=[round(float(v), 2) for v in fig.get_size_inches()],
                   n_text=len(texts), n_words=sum(len(t["text"].split()) for t in texts),
                   n_title=roles.count("title"), n_axis_label=roles.count("axis_label"), n_tick=roles.count("tick"),
                   n_legend=roles.count("legend"), n_annotation=len(ann),
                   n_numeric_annotation=sum(bool(re.search(r"\d", t["text"])) for t in ann),
                   annotation_words=sum(len(t["text"].split()) for t in ann),
                   n_overlap=int(n_overlap), min_font=min([t["size"] for t in texts], default=None),
                   n_panels=panels, n_series=series, n_bars=bars, n_ref_lines=ref_lines,
                   n_colors=len(colors), n_colors_nongrey=sum(not is_grey(h) for h in colors),
                   title_text=" | ".join(t["text"] for t in texts if t["role"] == "title")[:400],
                   texts=texts)
json.dump(out, open("metrics.json", "w", encoding="utf-8"))
'''


def render(code, brief_id, out_dir, timeout=180):
    """Run one chart's code. Writes code.py, chart.png and metrics.json to out_dir and returns the metrics."""
    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)
    code_sha = hashlib.sha256(code.encode("utf-8")).hexdigest()
    mpath = out_dir / "metrics.json"
    if mpath.exists():
        m = json.loads(mpath.read_text(encoding="utf-8"))
        if m.get("code_sha256") == code_sha and (not m.get("ok") or (out_dir / "chart.png").exists()):
            return m
    with tempfile.TemporaryDirectory(prefix="nb44run_") as tmp:
        shutil.copy(BRIEF_DIR / brief_id / "data.csv", Path(tmp) / "data.csv")
        Path(tmp, "chart_code.py").write_text(code, encoding="utf-8")
        Path(tmp, "runner.py").write_text(RUNNER, encoding="utf-8")
        try:
            p = subprocess.run([sys.executable, "-E", "-P", "-X", "utf8", "runner.py"], cwd=tmp, capture_output=True,
                               text=True, encoding="utf-8", errors="replace", timeout=timeout, env=_clean_env())
            mfile = Path(tmp, "metrics.json")
            m = json.loads(mfile.read_text(encoding="utf-8")) if mfile.exists() else {"ok": False, "error": (p.stderr or "runner crashed")[-1500:]}
        except subprocess.TimeoutExpired:
            m = {"ok": False, "error": f"Timed out after {timeout} s"}
        if m.get("ok"):
            shutil.copy(Path(tmp, "chart.png"), out_dir / "chart.png")
        elif (out_dir / "chart.png").exists():
            (out_dir / "chart.png").unlink()
    m["code_sha256"] = code_sha
    (out_dir / "code.py").write_text(code, encoding="utf-8", newline="\n")
    mpath.write_text(json.dumps(m, indent=1), encoding="utf-8", newline="\n")
    return m
````

````python
# Generation. For each brief, model and level: one call, render the code, and if the code fails to run,
# one repair call that shows the model its error (what a person would do). L4 starts from the model's own
# final L2 chart. A reply with no code in it is counted as unreadable and is not repaired.

def run_jobs(jobs, fn, workers=12):
    with ThreadPoolExecutor(workers) as pool:
        return list(pool.map(fn, jobs))


def produce(job):
    bid, mk, level = job
    b, model = BRIEF[bid], MAKERS[mk]
    out = CHART_DIR / bid / mk / level
    rec = {"brief": bid, "maker": mk, "level": level, "calls": [], "first_try_ok": False, "repaired": False}
    images = []
    if level == "L4":
        base = CHART_DIR / bid / mk / "L2"
        bm = json.loads((base / "metrics.json").read_text(encoding="utf-8")) if (base / "metrics.json").exists() else {}
        if not bm.get("ok"):
            rec.update(status="no_base", error="the L2 chart did not render, so there is nothing to revise")
            return rec
        prompt = revise_prompt(b, (base / "code.py").read_text(encoding="utf-8"))
        images = [base / "chart.png"]
    else:
        prompt = draft_prompt(b, level)
    row = ask(model, prompt, images, name=f"make {bid} {mk} {level}", session=f"{bid}/{mk}")
    rec["calls"].append(row["key"])
    resp = row["response"]
    if resp.get("error"):
        rec.update(status="call_failed", error=str(resp["error"])[:300])
        return rec
    code = extract_code(resp["text"])
    if code is None:
        rec.update(status="unreadable", error="reply had no python code block", reply_head=resp["text"][:300])
        return rec
    m = render(code, bid, out)
    rec["first_try_ok"] = bool(m.get("ok"))
    if not m.get("ok"):
        rec["first_error"] = m.get("error", "")[-600:]
        (out / "first_attempt.py").write_text(code, encoding="utf-8", newline="\n")
        row2 = ask(model, repair_prompt(prompt, code, m.get("error", "")[-1500:]), images,
                   name=f"repair {bid} {mk} {level}", session=f"{bid}/{mk}")
        rec["calls"].append(row2["key"])
        code2 = extract_code(row2["response"].get("text", ""))
        if code2 is not None:
            m = render(code2, bid, out)
            rec["repaired"] = bool(m.get("ok"))
    rec["status"] = "ok" if m.get("ok") else "code_failed"
    if not m.get("ok"):
        rec["error"] = m.get("error", "")[-600:]
    return rec


draft_jobs = [(b["id"], mk, lv) for b in BRIEFS for lv in ["L0", "L1", "L2", "L3"] for mk in MAKERS_USED]
draft_records = run_jobs(draft_jobs, produce)
revise_jobs = [(b["id"], mk, "L4") for b in BRIEFS for mk in MAKERS_USED]
revise_records = run_jobs(revise_jobs, produce)
records = pd.DataFrame(draft_records + revise_records)
print(records.groupby(["maker", "status"]).size().unstack(fill_value=0))
````

## 7. How a chart is measured

A chart is measured three ways, each answering a different question.

### Clutter, counted from the figure

After the code runs, the harness walks the matplotlib figure and lists every piece of text that is drawn, with its role: title, axis label, tick label, legend entry, or annotation. An annotation is any other text the code placed on the chart: a value printed on a bar, a note beside a line, a subtitle, a footnote. The study's main clutter measure is the number of annotations, called **labels** below, because labels are where charts most often say too much; titles, axis labels and tick labels are needed on almost every chart. The harness also counts the words on the chart, labels that contain a number, panels, data series, non-grey colours and pairs of text items that overlap. For scale, the twelve reference charts carry between 2 and 7 labels each.

### What a reader takes away: the quiz

Kimi K3.1 Pro sees only the chart image, never the data or the notes, and answers in two separate calls. The first call asks two open questions: what is the main message of the chart, in one sentence, and does the chart state any caveat about the data. Asking these first, in their own call, means the specific questions cannot hint at the answer. The second call asks for the two critical facts as numbers, for example "What was the Family plan's monthly cancellation rate before the April 2026 price change, and after it?", and allows `null` when the chart does not show a value.

The main-message answer is scored with a fixed pattern written for each brief before any chart was made: for B01 it must mention the Family plan together with cancellations rising or subscribers falling. The caveat answer is scored the same way: for B01 it must mention September together with "partial", "half", "1 to 14" or similar. Each numeric answer is correct when it is within a set tolerance of the value computed from the data (for B01, within 0.6 points of the 3.08% before the rise and 0.8 points of the 6.47% after). The **critical score** of a chart is the average of three parts: main message found (0 or 1), caveat handled (0 or 1), and the share of the critical numbers read correctly (0 to 1). A caveat counts as handled when the reader reports it, or when most judges say the chart left out the affected data so the caveat no longer applies.

### What a careful reviewer sees: the judge panel

Three judges, Claude Opus 5.5, GPT-6.1 Sol and Muse Wonder Full, rate every chart. Each judge sees the brief, the data, all ten notes marked CRITICAL, CAVEAT or SECONDARY, and the chart image, without knowing which model or level made it. For every note, the judge says whether the chart calls it out (states it or draws attention to it), shows it without pointing to it, or leaves it out. The count of secondary notes called out measures content overload, as opposed to the visual clutter that the label count measures. The judge also rates overload (1 = nothing competes with the main message, 5 = the message is hard to find), clarity (5 = the decision-critical message is obvious within ten seconds), lists errors such as wrong numbers or garbled text, and gives an overall readiness score from 1 to 10.

### A worked number

A small example of the critical score: a chart whose title says "Family cancellations doubled after the price rise" (main message found, 1), which shows September with no note (caveat not handled, 0), and from which the reader reads the before and after rates correctly but cannot read the subscriber counts (2 of 4 numbers, 0.5), scores (1 + 0 + 0.5) / 3 = 0.5. A chart scores 1.0 only if a reader who sees nothing but the chart gets the message, the caveat and every critical number.

The next three cells define the reader quiz, the judge prompt and the scoring loop.

````python
# The reader quiz. A neutral model (Kimi K3.1 Pro, from a vendor whose models make no charts here) sees
# only the chart image, never the data or notes. Call 1 asks the open questions first, so the specific
# questions cannot hint at the answer: what is the main message, and does the chart state a caveat?
# Call 2 asks for the two critical facts as numbers, scored in code against the answers computed from
# the data.

READER_OPEN = """The attached image is a chart shown at a business meeting. Answer from the image only.

1. In one sentence, what is the main message of this chart?
2. Does the chart state any caveat or limitation about the data (for example that some data are partial, missing, estimated or not comparable, or that something else changed at the same time)? If it does, describe it in one sentence. If it does not, write "none".

Reply with JSON only: {"main_message": "...", "caveat": "..."}"""


def reader_facts_prompt(b):
    qs = []
    for i, q in enumerate(b["quiz"]["facts"], 1):
        shape = ", ".join(f'"{f["name"]}": ...' for f in q["fields"])
        qs.append(f"{i}. {q['q']}\n   Answer as \"q{i}\": {{{shape}}}")
    return ("The attached image is a chart. Answer from the image only. Read values from labels if the chart prints them; "
            "otherwise estimate them from the axes. If the chart does not show a value at all, use null. "
            "Do not use outside knowledge and do not guess.\n\n" + "\n".join(qs) +
            "\n\nReply with JSON only. Give numbers without units or % signs, and names as text.")


def parse_json(text):
    """The first complete JSON object in a reply, or None."""
    text = text or ""
    dec = json.JSONDecoder()
    for i, ch in enumerate(text):
        if ch == "{":
            try:
                obj, _ = dec.raw_decode(text[i:])
                if isinstance(obj, dict):
                    return obj
            except ValueError:
                continue
    return None


def rule_hit(rule, text):
    t = (text or "").lower()
    if not t.strip() or t.strip() in ("none", "none.", "no", "n/a"):
        return False
    return any(all(re.search(rx, t) for rx in alt) for alt in rule["any"])


def to_number(v, ranges=True):
    """A number from a reply value. A range such as "17-27" counts as its midpoint; with ranges=False only
    the first number is read (the first version of this parser, kept for comparison)."""
    if isinstance(v, (int, float)) and not isinstance(v, bool):
        return float(v)
    if isinstance(v, str):
        s = v.replace("−", "-").replace(",", "")
        r = re.match(r"\s*(\d+\.?\d*)\s*%?\s*(?:-|–|to)\s*(\d+\.?\d*)", s)
        if ranges and r:
            return (float(r.group(1)) + float(r.group(2))) / 2
        m = re.search(r"-?\d+\.?\d*", s)
        if m:
            return float(m.group())
    return None


def field_correct(f, v, ranges=True):
    if v is None:
        return False
    if isinstance(f["answer"], str):
        return isinstance(v, str) and f["answer"].lower() in v.lower()
    x = to_number(v, ranges)
    if x is None:
        return False
    tol = f.get("tol", 0) if "rel" not in f else abs(f["answer"]) * f["rel"]
    return abs(x - f["answer"]) <= tol + 1e-9


def read_chart(b, png, tag):
    r1 = ask(READER, READER_OPEN, [png], name=f"read-open {tag}", session=tag)
    r2 = ask(READER, reader_facts_prompt(b), [png], name=f"read-facts {tag}", session=tag)
    o, f = parse_json(r1["response"].get("text")), parse_json(r2["response"].get("text"))
    res = {"reader_open_ok": o is not None, "reader_facts_ok": f is not None,
           "main_message": (o or {}).get("main_message"), "caveat_text": (o or {}).get("caveat")}
    res["main_hit"] = rule_hit(b["quiz"]["main"], res["main_message"])
    res["caveat_hit"] = rule_hit(b["quiz"]["caveat"], res["caveat_text"])
    n_ok, n_ok_first, n_all, n_null = 0, 0, 0, 0
    for i, q in enumerate(b["quiz"]["facts"], 1):
        ans = (f or {}).get(f"q{i}") or {}
        for fld in q["fields"]:
            v = ans.get(fld["name"]) if isinstance(ans, dict) else None
            n_all += 1
            n_null += v is None
            n_ok += field_correct(fld, v)
            n_ok_first += field_correct(fld, v, ranges=False)
    res.update(facts_correct=n_ok, facts_total=n_all, facts_null=n_null, fact_score=n_ok / n_all,
               fact_score_first_parser=n_ok_first / n_all,
               facts_raw=json.dumps((f or {}), ensure_ascii=False)[:600])
    return res
````

````python
# The judge panel. Three judges (Claude Opus 5.5, GPT-6.1 Sol, Muse Wonder Full) each see the brief, the
# data, every analyst note with its role, and the chart image. They do not know which model or level made
# the chart. Each judge marks how the chart treats every note and rates overload, clarity and readiness.

ROLE_LABEL = {"critical": "CRITICAL", "caveat": "CAVEAT", "secondary": "SECONDARY"}


def judge_prompt(b):
    csv_text = (BRIEF_DIR / b["id"] / "data.csv").read_text(encoding="utf-8").strip()
    notes = "\n".join(f"{f['id']} [{ROLE_LABEL[f['role']]}] {f['text']}" for f in b["findings"])
    keys = ", ".join(f'"{f["id"]}": "..."' for f in b["findings"])
    return f"""You are reviewing a chart made for a meeting. {b['audience'][0].upper() + b['audience'][1:]} must decide {b['decision']}. They will see the chart for about ten seconds.

The chart maker was given this data and these analyst notes. Each note is marked CRITICAL (needed for the decision), CAVEAT (changes how the data must be read) or SECONDARY (true, but not needed for this decision).

## Data: data.csv
```csv
{csv_text}
```

## Analyst notes
{notes}

The attached image is the chart. Judge only what the image shows.

For each note, say how the chart treats it:
- "called_out": the chart states it in text (title, subtitle, annotation or label) or draws attention to it (highlight colour, arrow, reference line or shading).
- "visible": the data behind it can be read from the chart, but nothing points to it.
- "absent": it cannot be read from the chart.

Then rate the chart:
- "overload" from 1 to 5: 1 = nothing competes with the main message; 3 = some clutter; 5 = so much text, labelling, series or colour that the main message is hard to find.
- "clarity" from 1 to 5: 5 = the decision-critical message is obvious within ten seconds; 1 = the reader cannot tell what matters.
- "caveat_handling": "stated" if the chart states the CAVEAT note; "avoided" if the chart leaves out the affected data so the caveat no longer applies; "missing" if the affected data is shown without the caveat.
- "errors": a list of factual errors or misleading choices (wrong numbers, mislabelled series, bar axes not starting at zero, garbled, overlapping or unreadable text). An empty list if there are none.
- "overall" from 1 to 10: how ready the chart is to show this audience as it is.

Reply with JSON only:
{{"notes": {{{keys}}}, "overload": 0, "clarity": 0, "caveat_handling": "...", "errors": [], "overall": 0, "reason": "one or two sentences"}}"""


def judge_chart(b, png, judge_key, tag):
    row = ask(JUDGES[judge_key], judge_prompt(b), [png], name=f"judge-{judge_key} {tag}", session=tag)
    j = parse_json(row["response"].get("text"))
    ok = (isinstance(j, dict) and isinstance(j.get("notes"), dict) and all(k in j for k in ("overload", "clarity", "overall")))
    res = {"judge": judge_key, "judge_ok": bool(ok)}
    if ok:
        notes = {f["id"]: str(j["notes"].get(f["id"], "")).lower() for f in b["findings"]}
        res.update(overload=to_number(j["overload"]), clarity=to_number(j["clarity"]), overall=to_number(j["overall"]),
                   caveat_handling=str(j.get("caveat_handling", "")).lower(),
                   n_errors=len(j.get("errors") or []), errors=json.dumps(j.get("errors") or [], ensure_ascii=False)[:800],
                   reason=str(j.get("reason", ""))[:400], notes=json.dumps(notes))
    return res
````

````python
# Score every rendered chart: the reader quiz (two calls) and one call per judge. Jobs from all backends
# are interleaved so that the slow ones (Muse, the reader) do not idle the fast ones.
charts = []
for b in BRIEFS:
    charts.append({"brief": b["id"], "maker": "reference", "level": "ref", "png": CHART_DIR / b["id"] / "reference" / "chart.png"})
    for mk in MAKERS_USED:
        for lv in LEVELS:
            p = CHART_DIR / b["id"] / mk / lv
            if (p / "metrics.json").exists() and json.loads((p / "metrics.json").read_text(encoding="utf-8")).get("ok"):
                charts.append({"brief": b["id"], "maker": mk, "level": lv, "png": p / "chart.png"})


def tag_of(c):
    return f"{c['brief']} {c['maker']} {c['level']}"


def score_job(job):
    kind, c = job
    b = BRIEF[c["brief"]]
    if kind == "reader":
        return ("reader", c, read_chart(b, c["png"], tag_of(c)))
    return ("judge", c, judge_chart(b, c["png"], kind, tag_of(c)))


score_jobs = [(k, c) for c in charts for k in ["reader"] + JUDGES_USED]
score_out = run_jobs(score_jobs, score_job, workers=16)
key_cols = lambda c: {"brief": c["brief"], "maker": c["maker"], "level": c["level"]}
reader_df = pd.DataFrame([{**key_cols(c), **r} for kind, c, r in score_out if kind == "reader"])
judge_df = pd.DataFrame([{**key_cols(c), **r} for kind, c, r in score_out if kind == "judge"])
print("Charts scored:", len(charts), "| reader replies parsed:", int(reader_df.reader_open_ok.sum() + reader_df.reader_facts_ok.sum()),
      "of", 2 * len(reader_df), "| judge replies parsed:", int(judge_df.judge_ok.sum()), "of", len(judge_df))
````

## 8. Results tables and a worked example

The cell below joins everything into one row per chart: the counts from the figure, the reader's answers and scores, and the judges' ratings averaged over the three judges. It writes the tables to `data/results/` and prints the means by model and level. The functions after it compare two levels on the same briefs: for each brief and model, the difference between two levels; then the mean of those differences, with a 95% interval from resampling the twelve briefs (the brief, not the chart, is the unit that varies).

````python
# One row per chart: generation outcome, clutter counts from the figure, the reader's results and the
# judges' ratings averaged over the judges. Tables are written to data/results/.
RESULTS = DATA / "results"
RESULTS.mkdir(exist_ok=True)
MCOLS = ["n_text", "n_words", "n_title", "n_axis_label", "n_tick", "n_legend", "n_annotation", "n_numeric_annotation",
         "annotation_words", "n_overlap", "n_panels", "n_series", "n_bars", "n_colors", "n_colors_nongrey", "title_text"]


def garbled_text(s):
    """matplotlib pairs unescaped $ signs and renders the text between them as maths, dropping the $ signs and
    italicising the words. A pair whose content starts with a digit is a currency amount eaten this way."""
    idx = [i for i, ch in enumerate(s) if ch == "$" and (i == 0 or s[i - 1] != "\\")]
    return any(s[a + 1:a + 2].isdigit() for a, b in zip(idx[0::2], idx[1::2]))


def dollar_garble(m):
    return any(garbled_text(t["text"]) for t in m.get("texts", []))


def free_text_title(m):
    """1 if the chart's title was drawn as free text (fig.text or ax.text) rather than with set_title or suptitle:
    the topmost text item is an annotation at least as large as any proper title. That item is a title, not a label."""
    texts = m.get("texts", [])
    if not texts:
        return 0
    top = max(texts, key=lambda t: t["bbox"][3])
    title_sizes = [t["size"] for t in texts if t["role"] == "title"]
    return int(top["role"] == "annotation" and top["size"] >= max(title_sizes, default=0))


gen = records.drop(columns=["calls"]).copy()
gen["n_calls"] = records.calls.apply(len)
met_rows = []
for c in charts:
    m = json.loads((c["png"].parent / "metrics.json").read_text(encoding="utf-8"))
    met_rows.append({"brief": c["brief"], "maker": c["maker"], "level": c["level"], **{k: m.get(k) for k in MCOLS},
                     "dollar_garble": dollar_garble(m), "free_text_title": free_text_title(m)})
met = pd.DataFrame(met_rows)

ROLE = {b["id"]: {f["id"]: f["role"] for f in b["findings"]} for b in BRIEFS}


def judge_counts(r):
    if not r["judge_ok"]:
        return pd.Series(dtype=float)
    notes = json.loads(r["notes"])
    roles = ROLE[r["brief"]]
    called = [k for k, v in notes.items() if v.startswith("called")]
    shown = [k for k, v in notes.items() if v.startswith("called") or v.startswith("visible")]
    return pd.Series({"crit_called": sum(roles[k] == "critical" for k in called),
                      "sec_called": sum(roles[k] == "secondary" for k in called),
                      "sec_shown": sum(roles[k] == "secondary" for k in shown),
                      "cav_stated": float(r["caveat_handling"].startswith("stated")),
                      "cav_avoided": float(r["caveat_handling"].startswith("avoided"))})


jc = judge_df.join(judge_df.apply(judge_counts, axis=1))
jmean = jc[jc.judge_ok].groupby(["brief", "maker", "level"])[["crit_called", "sec_called", "sec_shown", "cav_stated", "cav_avoided",
                                                               "overload", "clarity", "overall", "n_errors"]].mean().reset_index()
rd = reader_df.copy()
chart_df = met.merge(rd, on=["brief", "maker", "level"], how="left").merge(jmean, on=["brief", "maker", "level"], how="left")
# A caveat counts as handled when the reader found it on the chart, or most judges say the chart left out the
# affected data so the caveat no longer applies.
chart_df["caveat_ok"] = (chart_df.caveat_hit | (chart_df.cav_avoided > 0.5)).astype(float)
chart_df["critical_score"] = (chart_df.main_hit.astype(float) + chart_df.caveat_ok + chart_df.fact_score) / 3
chart_df["critical_score_first_parser"] = (chart_df.main_hit.astype(float) + chart_df.caveat_ok + chart_df.fact_score_first_parser) / 3
chart_df["labels"] = chart_df.n_annotation - chart_df.free_text_title
model_df = chart_df[chart_df.maker != "reference"].copy()
ref_df = chart_df[chart_df.maker == "reference"].copy()

for name, df in [("generation", gen), ("charts", chart_df.drop(columns=["facts_raw"])), ("reader", reader_df), ("judges", judge_df)]:
    df.to_csv(RESULTS / f"{name}.csv", index=False)


def boot_diff(df, metric, a, b, by_maker=None, n=4000, seed=44):
    """Mean of (level b minus level a) over brief x model pairs, with a 95% interval from resampling briefs."""
    d = df if by_maker is None else df[df.maker == by_maker]
    p = d.pivot_table(index=["brief", "maker"], columns="level", values=metric).dropna(subset=[a, b])
    diff = (p[b] - p[a]).groupby(level="brief").mean()
    rng = np.random.default_rng(seed)
    briefs = diff.index.to_numpy()
    sims = [diff.loc[rng.choice(briefs, len(briefs))].mean() for _ in range(n)]
    return {"metric": metric, "from": a, "to": b, "maker": by_maker or "all", "pairs": len(p),
            "mean_diff": diff.mean(), "ci_low": np.percentile(sims, 2.5), "ci_high": np.percentile(sims, 97.5)}


level_table = model_df.groupby(["maker", "level"]).agg(
    charts=("brief", "size"), labels=("labels", "mean"), numbers=("n_numeric_annotation", "mean"), words=("n_words", "mean"),
    panels=("n_panels", "mean"), sec_called=("sec_called", "mean"), main_hit=("main_hit", "mean"), caveat_ok=("caveat_ok", "mean"),
    facts=("fact_score", "mean"), critical=("critical_score", "mean"), overload=("overload", "mean"), clarity=("clarity", "mean"),
    overall=("overall", "mean")).round(2)
display(level_table)
````

````python
LV = ["L0", "L1", "L2", "L3", "L4"]
LV_TICKS = ["L0\nbare", "L1\npurpose", "L2\nprinciples", "L3\npriorities", "L4\nrevise"]


def level_means(metric, df=None):
    d = model_df if df is None else df
    return d.pivot_table(index="level", columns="maker", values=metric, aggfunc="mean").reindex(LV)


def spread_labels(values, gap):
    """Nudge end-of-line label positions apart so that no two are closer than gap (data units)."""
    order = sorted(values, key=values.get)
    pos, last = {}, None
    for k in order:
        y = values[k] if last is None else max(values[k], last + gap)
        pos[k], last = y, y
    return pos


def line_panel(ax, metric, ylabel, ref_value=None, ylim=None, fmt="{:.1f}"):
    t = level_means(metric)
    for mk in MAKERS_USED:
        ax.plot(range(5), t[mk], color=COLORS[mk], lw=2.2, marker="o")
    if ylim:
        ax.set_ylim(*ylim)
    lo, hi = ax.get_ylim()
    pos = spread_labels({mk: t[mk].iloc[-1] for mk in MAKERS_USED}, 0.055 * (hi - lo))
    for mk in MAKERS_USED:
        ax.text(4.12, pos[mk], f"{MAKERS[mk]['label']}  " + fmt.format(t[mk].iloc[-1]), color=COLORS[mk], va="center", fontsize=9)
    if ref_value is not None:
        ax.axhline(ref_value, color=COLORS["reference"], ls="--", lw=1.2)
        ax.text(0, ref_value + 0.012 * (hi - lo), "reference charts " + fmt.format(ref_value), color=COLORS["reference"], fontsize=8.5, va="bottom")
    ax.set_xticks(range(5), LV_TICKS)
    ax.set_xlim(-0.2, 5.6)
    ax.set_ylabel(ylabel)
````

## 11. Results: critical information a reader takes away

A chart with few labels is only useful if the critical information is still on it. The right panel of the figure in section 10 shows the critical score: the average of three parts, each scored from the image alone by a reader model that never saw the data. The figure below takes the score apart. The left panel shows how often the reader found the main message and how often the caveat was handled. The right panel splits the 132 critical numbers asked about at each level (36 charts, three or four numbers each) into numbers the reader read correctly, numbers it read wrongly, and numbers that were not on the chart, where it answered `null`.

````python
# Where the critical score comes from: main message, caveat, and the critical numbers, split into numbers read
# correctly, numbers the chart did not show (the reader answered null) and numbers read wrongly.
parts = model_df.groupby("level").agg(main=("main_hit", "mean"), caveat=("caveat_ok", "mean"), correct=("facts_correct", "sum"),
                                      missing=("facts_null", "sum"), total=("facts_total", "sum")).reindex(LV)
parts["wrong"] = parts.total - parts.correct - parts.missing
fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw={"width_ratios": [1, 1.25]})
x = np.arange(5)
a1.plot(x, 100 * parts.main, marker="o", color="#333", lw=2)
a1.plot(x, 100 * parts.caveat, marker="s", color="#8E44AD", lw=2)
a1.text(4.1, 100 * parts.main.iloc[-1] + 1.5, "main message found", fontsize=9, color="#333")
a1.text(4.1, 100 * parts.caveat.iloc[-1] - 4.5, "caveat handled", fontsize=9, color="#8E44AD")
a1.set_ylim(50, 102)
a1.set_xlim(-0.2, 5.4)
a1.set_xticks(x, LV_TICKS)
a1.set_ylabel("% of 36 charts")
a1.set_title("The message and the caveat come through at every level", fontsize=11)
bottom = np.zeros(5)
for col, colr, lab in [("correct", "#2E8B57", "read correctly"), ("wrong", "#E67E22", "read wrongly"), ("missing", "#BDC3C7", "not on the chart")]:
    vals = 100 * parts[col] / parts.total
    a2.bar(x, vals, bottom=bottom, color=colr, width=0.62)
    for i, v in enumerate(vals):
        if v >= 7:
            a2.text(i, bottom[i] + v / 2, f"{v:.0f}%", ha="center", va="center", fontsize=9, color="white" if col != "missing" else "#333")
    a2.text(4.38, bottom[-1] + vals.iloc[-1] / 2, lab, va="center", fontsize=9, color=colr if col != "missing" else "#666")
    bottom += vals.to_numpy()
a2.set_xticks(x, LV_TICKS)
a2.set_xlim(-0.5, 5.4)
a2.set_ylim(0, 100)
a2.set_ylabel(f"% of the {int(parts.total.iloc[0])} critical numbers asked about")
a2.set_title("The critical numbers vanish under general design advice", fontsize=11)
fig.tight_layout()
fig.savefig(FIG / "critical_parts.png")
plt.show()
display(parts)
````

The main message and the caveat come through almost everywhere, so they cannot rank the levels. The reader found the main message on all 36 bare-request charts and on 89% to 100% at the other levels; the caveat was handled on 89% of bare-request charts and 94% to 97% at the other levels. Both sit near the ceiling, and the few misses are discussed as individual cases in section 13.

The critical numbers are where the levels differ. With a bare request the reader read 84% of them correctly. Under the design rules (L2) it read 57%, and the drop is almost entirely numbers that were no longer on the chart: 32% of the L2 numbers drew a `null`, against 9% at L0. The rule "add a number label only where the number is part of the message" was read as "print fewer numbers", and the numbers dropped were often the ones the decision needed, such as the cancellation rates before and after the price rise or the share of loan value that comes from brokers. The size of the drop differs by model. DeepSeek V4.1 Flash fell from 83% to 36% of numbers read, Muse Wonder Full from 94% to 61%, and Claude Opus 5.5 hardly changed (75% to 72%).

The critical score rose from 0.81 at L2 to 0.98 at L3 (difference 0.17, interval 0.12 to 0.23), close to the reference charts' 1.00.

The critical score in this section uses a parser fixed after the results were seen: ten answers to one question in B12 gave the range printed on the chart ("17-27") rather than a single number, and the first parser read only the first number. The fixed parser reads a range as its midpoint. ### A bug that raises no error: dollar signs read as maths

In matplotlib, text between two dollar signs is drawn as a mathematical formula: the signs disappear and the words become italic. A label written as "\$79 to \$89" is drawn as "79to89", and "\$3.1M of \$4.8M" as "3.1Mof4.8M". The code runs without any error or warning, so a harness that only checks that the code runs cannot catch it. 40 of the 180 model charts (22%) contain at least one such label: 8 at L0, 12 at L1, 8 at L2, 9 at L3 and 3 at L4. It appears in 7 of the 12 briefs, mostly where the notes talk about money: B01 (prices), B04 (budgets) and B11 (loan costs) account for 33 of the 40. None of the models avoided it reliably. The study author made the same mistake in the first drafts of three reference charts and fixed it after looking at them.

Looking at the rendered chart is what catches it. Claude Opus 5.5 had garbled text in 3 of its L2 charts and in none of the L4 revisions of them: after one look, it escaped the dollar signs (figure below, B01). Across the three models, the revision pass fixed 5 of the 8 garbled L2 charts. At least one judge noticed the problem on 26 of the 40 garbled charts, and all three on 10. Claude Opus 5.5 mentioned it in 60% of its judgements of those charts, Muse Wonder Full in 42% and GPT-6.1 Sol in 28%. The revision pass also cut the errors the judges listed per chart from 0.89 at L2 to 0.39 at L4.

````python
panel_grid([("B01", "opus", "L2", "'$79 to $89' rendered as maths: the dollar signs vanish"),
            ("B01", "opus", "L4", "after one look at the chart, the text is escaped")],
           ncols=2, path=FIG / "example_revise.png", width=7.2)
g = model_df.pivot_table(index="maker", columns="level", values="dollar_garble", aggfunc="sum")
g.index = [MAKERS[m]["label"] for m in g.index]
display(g)
garbled = model_df[model_df.dollar_garble][["brief", "maker", "level"]]
seen = jc.merge(garbled, on=["brief", "maker", "level"])
seen["noticed"] = seen.errors.fillna("").str.contains(r"dollar|math|garbl|italic|latex|\$ sign", case=False, regex=True)
print(f"Garbled charts: {len(garbled)}; noticed by at least one judge: {int(seen.groupby(['brief', 'maker', 'level']).noticed.any().sum())}; "
      f"share of judgements that mention it: " + ", ".join(f"{JUDGES[k]['label']} {v:.0%}" for k, v in seen.groupby('judge').noticed.mean().items()))
fixed = model_df.pivot_table(index=["brief", "maker"], columns="level", values="dollar_garble")
print("Garbled at L2 and fixed at L4:", int(((fixed["L2"] == 1) & (fixed["L4"] == 0)).sum()), "of", int((fixed["L2"] == 1).sum()))
````

## 14. Limitations

The reader is a model, not a person in a meeting. Kimi K3.1 Pro reads every label at full resolution, and it found the main message on all 36 bare-request charts, including the one with 118 labels. A person looking at a slide for ten seconds would very likely do worse on crowded charts, so the critical score probably flatters the bare-request and purpose levels. The judges' overload and clarity ratings capture some of this, but no person read these charts under time pressure. The comparison between L2 and L3 is less exposed to this problem, because both levels produce charts with few labels.

Two of the three judges are also chart makers. Claude Opus 5.5 and Muse Wonder Full each judged their own charts along with the others'. Section 12 shows no sign that either favoured its own: GPT-6.1 Sol, whose vendor makes no charts here, rated every model's charts higher than both of them did, and all three judges ranked the makers in the same order. The judges agree on clutter but not on whether a clean drawing makes up for missing numbers: Sol preferred the L3 charts to the L4 charts and the other two did not, so a judge-based verdict on that question depends on the judge chosen.

Each brief, model and level got one chart. Models give different charts on different runs, so a single chart per cell mixes the effect of the level with run-to-run variation. The intervals in this notebook resample briefs, so they reflect how much the results vary across situations, not how much they would vary if the same prompt were sent again.

The briefs are tidy. Each has a small table, ten notes that an analyst has already written, and exactly two critical facts and one critical caveat, defined by the study. Real requests are messier: no notes, more data, and a decision that is less clearly stated. The study's definition of "critical" is also one reading of each decision. In section 13, two of Claude Opus 5.5's charts (B08 and B11) chose a different but defensible message and scored low for it.

The design rules at L2 are one wording. The rule "add a number label only where the number is part of the message" is the most likely cause of the missing numbers at L2, and a different wording might keep them. Section 16 proposes testing that directly.

All charts are matplotlib code. The dollar-sign problem is specific to matplotlib, and the label counts depend on how matplotlib represents text. A chart whose title was drawn as free text rather than as a title is counted with that item as a title, not a label; this rule was added after the first look at the results. Without it the label counts by level are 15.4, 12.5, 5.6, 7.9 and 6.2 instead of 15.4, 12.1, 5.3, 7.4 and 5.6, and no comparison changes.

The scoring was fixed before the run except for two changes made after seeing the results, both reported with the numbers under the original rule: the free-text title rule above and the range parser in section 11. The main-message and caveat patterns were not changed; a person reading the seven summaries that missed the pattern would accept two of them (section 13). The full run history is in [`data/PROTOCOL.md`](data/PROTOCOL.md).

### Further reading

Links checked on 5 November 2026.

- [MatPlotAgent: Method and Evaluation for LLM-Based Agentic Scientific Data Visualization](https://arxiv.org/abs/2402.11453) (Yang et al., 2024). An agent that writes plotting code, renders it and revises it from a vision model's feedback; level L4 here is one round of the same idea, and it fixed the drawing more than the content.
- [VisEval: A Benchmark for Data Visualization in the Era of Large Language Models](https://arxiv.org/abs/2407.00981) (Chen et al., 2024). A benchmark that checks charts generated by language models for validity, legality and readability, partly with a vision model, much as this notebook combines counts with judges.
- [ChartQA: A Benchmark for Question Answering about Charts with Visual and Logical Reasoning](https://arxiv.org/abs/2203.10244) (Masry et al., 2022). Questions answered from chart images alone; the reader quiz here applies the same idea to the study's own charts, with answers computed from the data.
- [Graphical Perception: Theory, Experimentation, and Application to the Development of Graphical Methods](https://doi.org/10.1080/01621459.1984.10478080) (Cleveland and McGill, 1984). Experiments on how accurately people read values from different chart encodings, the background for preferring positions on a common scale to colour and area.
- [What Makes a Visualization Memorable?](https://doi.org/10.1109/TVCG.2013.234) (Borkin et al., 2013). Found that charts with more colour and recognisable elements are remembered better, a reminder that what is memorable and what serves a quick decision are different targets.
- [Frames and Slants in Titles of Visualizations on Controversial Topics](https://doi.org/10.1145/3173574.3174012) (Kong, Liu and Karahalios, 2018). Showed that a chart's title shapes what readers take away from it, the reason the design rules here ask for a title that states the finding.
