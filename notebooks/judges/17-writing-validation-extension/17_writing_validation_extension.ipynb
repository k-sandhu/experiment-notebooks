# Does a writer's rank depend on which AI judges score it?

## Comparing AI writers with AI judges

A team that wants an AI model to draft its reports has to choose one, and the obvious way is to give several models the same tasks and compare the results. The hard part is the comparison. A person can read and score a few dozen documents carefully, but not several hundred, and not again every time a new model comes out. So teams often ask other AI models to do the scoring. These models are **judges**: each reads a document and returns ratings, such as how clear it is and whether its facts are right, and sometimes a list of problems.

Judges are fast, but they are models too, with tastes of their own. One worry is favouritism: a judge may score writing from its own company's models more kindly. A study of this found that models do prefer their own writing ([Panickssery et al., 2024](https://arxiv.org/abs/2404.13076)). A common precaution follows from it: never let a judge score a writer from its own company. An OpenAI model's drafts are scored only by judges from other companies, an Anthropic model's drafts likewise, and so on.

### An illustrative example

This example is made up, with easy numbers.

Two companies each make one writer and one judge. Judge A, from company A, is generous: it gives about 95 out of 100 to any clear, correct draft. Judge B, from company B, is strict: it gives about 87 to the same drafts. Now suppose writer A and writer B write equally well.

Under the precaution, writer A may only be scored by judge B, so it gets about 87. Writer B may only be scored by judge A, so it gets about 95. Writer B comes out 8 points ahead, and nothing about the writing differs. The precaution removed one possible bias, favouritism, and let in another, the difference in strictness between the two judges. Each writer's score now says as much about the judge it drew as about its writing.

The other way is to give every writer the same judges, its own company's judge included, and weight the two companies equally. Both writers then get (95 + 87) / 2 = 91, and the tie shows. Favouritism could still creep in, but it now has to show up as a gap that the strictness difference does not explain, which can be checked.

### Judges also point at problems

Judges are also used to find faults: a team asks for a list of weak passages and edits them. An earlier study in this repository ([notebook 15](../15-analytical-writing-style/)) asked 27 AI models to write 213 short decision documents on two invented cases, and asked AI judges to score them. When judges were told to hunt for "AI slop" (padded, formulaic or empty writing) and given eight habits to count, their clarity scores fell by 1.66 points out of 100 on the same drafts. But that instruction changed two things at once, the label "AI slop" and the list of habits. And two judges reading the same draft often flagged different passages: on one article, one judge flagged 1 passage and another flagged 13. Nobody checked whether those flags would survive a second, independent reading.

### What this study does

This study does three things. Six writers produce 288 longer decision documents on 24 new cases, three judges score all of them, and the writers' average scores are compared under the two ways of choosing judges: the **outside-company panel**, which drops same-company judges, and the **common panel**, which gives every writer the same three judges. Fresh judges re-read 204 passages from the earlier study, some that it flagged and some that it did not, without being told which. And the same 24 drafts are scored with a fixed list of the eight habits, with and without a warning that the text "may contain AI-slop prose", to separate the label from the list.

## The question

Does the ranking of six AI writers on 288 new decision documents depend on which AI judges are allowed to score each writer? And do AI judges' criticisms of prose hold up: when fresh judges re-read 204 passages from an earlier study, do they agree with its flags, and does warning a judge about "AI slop" change its scores when its criteria stay the same?

## Study summary

Six writers (GPT-6 Astra; Claude Opus 5.5, Sonnet 5.5 and Haiku 5.5; Muse Wonder Full; DeepSeek V4 Pro) each wrote two drafts for each of 24 new cases, 288 drafts in all, every one usable. Three judges scored every draft: two from OpenAI (GPT-6 Astra and GPT-6.1 Sol) and one from Anthropic (Claude Opus 5.5). The main measure is **clarity**, a 0 to 100 rating of how easy the analysis is to follow. 282 drafts have readable scores from all three judges.

- The ranking depended on the panel. Dropping same-company judges put GPT-6 Astra last of six, at 88.21. Giving every writer the same three judges, with the two companies weighted equally, put it first, at 93.10. No text changed.
- The cause is a gap in strictness, not favouritism. On every one of the 282 drafts, the two OpenAI judges' average clarity was higher than the Claude judge's, by 9.01 points on average (3 to 24). Dropping same-company judges left Astra with only the strict Claude judge and the Claude writers with only the generous OpenAI judges. Scored by the Claude judge alone, Astra was ahead of every Claude writer (88.21 against 87.92 for Claude Opus 5.5).
- The earlier study's prose flags rarely held up. On 96 previously flagged passages, fresh judges called the wording a problem in 1.6% of their opinions, against 1.2% on 108 unflagged passages. They raised some concern more often on flagged passages (16.1% against 5.2%), but 29 of those 38 concerns were about reasoning, mostly a claim the evidence could not support, not about wording.
- With the habit list fixed, adding the AI-slop warning changed clarity by -0.31 (95% interval -0.70 to +0.02) on 55 judge-and-draft pairs, and left 32 of the 55 scores exactly where they were.

Unreadable replies are counted apart from scores: 3 of 432 scoring requests broke the reply format and are left out. All 288 drafts arrived, though 76 missed their requested length.

For a team choosing a writer: give every writer the same judges and report each judge's scores separately. A rule that drops same-company judges can end up ranking writers by which judge they drew.

The main caveat is that every judgment here comes from a model. Forms for a study with human readers and editors are ready, but no responses have been collected, so nothing here shows which ranking or which criticism human readers would agree with.

One draft travels through it like this. Claude Opus 5.5 receives the `F1-1` email prompt and writes a draft. GPT-6 Astra, GPT-6.1 Sol and Claude Opus 5.5 each score it in a fresh session. Its outside-company score uses only the two OpenAI judges; its common-panel score also uses the Claude judge. Because it was also drawn for case `F1-1` in the warning test, the same draft is scored three more times by each judge, once under each instruction. Section 6 follows it step by step.

## 4. The 24 new cases

Each new case is an evidence packet with a decision rule and an exact answer, generated in code before any writer is called. The 24 cases cover six **decision types**, four parameter draws each:

| Decision type | What the writer must work out |
|---|---|
| Value threshold | Whether the added value of a programme exceeds its implementation cost and vendor fee |
| Supplier choice | The cheapest option that meets peak capacity and the budget |
| Changing mix | The real gain once a shift between standard and complex work is taken out of the raw total |
| Capacity bottleneck | Which stage limits completed work, and which option adds the most net value |
| Uncertain scenarios | Whether the expected net value is positive and the low scenario's loss stays within a limit |
| Trial with guardrail | Whether a randomised trial clears all three limits: success gain, rise in adverse events and net value |

Varying the decision type gives writers different reasoning problems, not just new names and numbers. Each case is assigned one document type in rotation, so each type gets one case of each decision type, and genre and case are not crossed: differences between document types also reflect different cases. The requested lengths are an email of 180 to 280 words, text for six slides of 200 to 320 words, a web article of 700 to 1,000 words and an academic-style report of 1,000 to 1,400 words. The packets leave out the licence and source notes that the earlier study included, after writers there copied the phrase "seeded arithmetic fixture" into their drafts.

## 6. How the judges scored the drafts

Every usable draft goes to GPT-6 Astra, GPT-6.1 Sol and Claude Opus 5.5, two drafts per request, each request in a fresh session with the author hidden. The instruction contains no mention of AI slop.

### The measures

Each judge gives four 0 to 100 ratings, higher being better. **Clarity** asks how easy the analysis is to follow and is the main measure. **Factual support** asks whether claims match the evidence (`grounding` in the files). **Reasoning** asks whether the conclusion follows. **Fit to the document type** asks whether it works as the requested email, slides, article or report (`genre_fit`). These are judges' ratings, not shares of readers who understood the text. A judge may also list up to three **concerns**, each quoting the text and typed as prose, factual, reasoning or format, and may list none; three is a cap, so the lists are not complete counts.

### The two panels

A writer's **panel** is the set of judges whose scores enter its average. Scores are first averaged over a writer's two drafts within each case, then over its 24 cases.

- The **outside-company panel** drops same-company judges. Claude writers are scored by the two OpenAI judges, GPT-6 Astra by the Claude judge alone, and Muse Wonder Full and DeepSeek V4 Pro by all three, averaged equally. It was the planned main analysis. The files call it `outside_vendor`.
- The **common panel** gives every writer all three judges: the two OpenAI judges are averaged first, and that average and the Claude judge get equal weight. It was added during analysis as a check (section 10). The files call it `common_vendor_balanced`.

Intervals are 95% intervals from resampling the 24 cases 1,000 times; a second interval resamples the six decision types instead.

## 7. How the earlier passages were re-checked

Each fresh judge sees the earlier draft and its task, with one passage marked, and answers: does this passage have a justified local problem for its reader, of which type (prose, factual, reasoning or format), and how severe? Severity is 0 for no justified problem, 1 for a minor nuisance, 2 for something a reader would likely reread and 3 for a likely misunderstanding. The instruction says that needed technical terms, justified caveats, headings, useful summaries and politeness can be appropriate, and asks the judge to keep prose difficulty apart from factual and reasoning errors.

The judge is not told the author, the old judge, the old flag or whether the passage was flagged. The writer's own model and the model that made the original flag are left out of each passage's judges, leaving two or three per passage. Up to four passages share a request. The result for a group is the share of judge opinions finding a problem, averaged within each passage and then over passages. Neither group has a correct answer, so the results are rates of agreement, not accuracy.

## 9. Materials for a human study

The question behind every model judgment here is whether people would agree, so the notebook builds the materials for a human study. No responses have been collected, and no model or test response is entered as a human one.

The reader page shows each reader one version of each case, with no author or answer key. A reader opens a document for a fixed time (90 seconds for an email, 120 for slides, 300 for an article, 420 for a report), the text is hidden, and four questions follow, with confidence and effort ratings. Accuracy is the main outcome; time and effort come second. The editor page gives editors the marked passages with their full drafts and evidence, without old flags or authors, and asks them to classify each as a prose, factual, reasoning or format problem, or no problem. ## 10. How the study got here

This study was designed after the earlier one, and four choices come from what that study showed. Its reading check, in which models answered questions from each draft, was correct on every readable reply and so could not tell drafts apart; this study prepares a human reading test instead (section 9). Its single short length target suited emails but not reports, so each document type here has its own range. Its slop audit changed the label and the list together, so the warning test separates them (section 8). And its packets carried a source note that writers copied into their text, so the new packets leave it out.

The protocol and the 24 cases were saved in the repository before any draft or score was collected, with the outside-company panel as the main analysis. Two choices came later. While replies were coming in, the repair for replies missing only their closing brackets was recorded (section 2); it affected 51 main scoring replies and 27 warning-test replies, all from the OpenAI judges, and every result is also given without them. And the common panel is not named in the protocol: it was computed alongside the first results as a check, and the breakdown by judge company that explains it (section 11) was added two days later. ## 13. Whether an AI-slop warning changes scores

````python
fw=framing[framing.external].pivot_table(index=['doc_id','case_id','judge','model'],columns='condition',values='clarity').dropna(subset=['baseline','rubric','warning'])
effects=[]
for name,left,right in [('rubric_minus_baseline','rubric','baseline'),('warning_minus_rubric','warning','rubric')]:
 delta=(fw[left]-fw[right]).groupby(level='case_id').mean();lo,hi=bootstrap(delta,name)
 family_delta=delta.groupby(delta.index.map(lambda cid:CASE_MAP[cid]['family'])).mean();family_lo,family_hi=bootstrap(family_delta,name+'-family')
 effects.append(dict(effect=name,mean=delta.mean(),lo=lo,hi=hi,family_lo=family_lo,family_hi=family_hi,cases=len(delta),judge_document_pairs=len(fw)))
effect_table=pd.DataFrame(effects);effect_table.to_csv(DATA/'instruction_effects.csv',index=False);fw.reset_index().to_csv(DATA/'matched_instruction_ratings.csv',index=False)
display(effect_table.round(3))
fig,ax=plt.subplots(figsize=(9,3.2))
labels=['Neutral rubric minus baseline','AI-slop warning minus identical rubric']
ax.errorbar(effect_table['mean'],labels,xerr=[effect_table['mean']-effect_table.lo,effect_table.hi-effect_table['mean']],fmt='o',capsize=4)
ax.axvline(0,color='#777',lw=1);ax.set(xlabel='Paired clarity change /100',title='Separate the detailed rubric from its warning label');ax.invert_yaxis()
fig.tight_layout();fig.savefig(FIG/'framing.png');plt.show()
print('Human reader responses:',len(reader_rows),'; human editor responses:',len(editor_rows),'. Pending means no human effect estimate is available.')
alignment=pd.DataFrame(columns=['doc_id','accuracy','clarity','grounding','reasoning','genre_fit'])
editor_alignment=pd.DataFrame(columns=['item_id','human_prose_concern','model_prose_concern','editors','model_judges'])
if reader_rows:
 hr=pd.DataFrame(reader_rows);ha=hr[hr.arm=='fresh'].groupby('doc_id').accuracy.mean().reset_index()
 model_doc=doc_scores.groupby('doc_id')[['clarity','grounding','reasoning','genre_fit']].mean().reset_index()
 alignment=ha.merge(model_doc,on='doc_id')
 if len(alignment)>=3:
  associations={dim:alignment[dim].rank().corr(alignment.accuracy.rank()) for dim in ['clarity','grounding','reasoning','genre_fit']}
  print('Descriptive rank associations with human accuracy:',associations,'on',len(alignment),'documents; shared cases and limited coverage prevent a population claim.')
  fig,ax=plt.subplots(figsize=(7,4));ax.scatter(alignment.clarity,alignment.accuracy*100)
  ax.set(xlabel='External model clarity /100',ylabel='Human answer accuracy (%)',ylim=(0,100),title='Observed reader accuracy versus model ratings')
  fig.tight_layout();fig.savefig(FIG/'human_model_alignment.png');plt.show()
if editor_rows:
 he=pd.DataFrame(editor_rows);he['prose_concern']=he.problem & (he.type=='prose')
 eh=he.groupby('source_item_id').agg(human_prose_concern=('prose_concern','mean'),editors=('person','nunique')).reset_index()
 eh=eh[eh.editors>=2]
 ma=adj.groupby('item_id').agg(model_prose_concern=('prose_problem','mean'),model_judges=('judge','nunique')).reset_index()
 editor_alignment=eh.merge(ma,left_on='source_item_id',right_on='item_id').drop(columns='source_item_id')
 print('Independent editor-model overlap:',len(editor_alignment),'passages. Editor disagreement remains in the proportions.')
alignment.to_csv(HUMAN/'reader_model_alignment.csv',index=False)
editor_alignment.to_csv(HUMAN/'editor_model_alignment.csv',index=False)

by_judge=fw.copy();by_judge['rubric_minus_baseline']=by_judge.rubric-by_judge.baseline;by_judge['warning_minus_rubric']=by_judge.warning-by_judge.rubric
by_judge.groupby(level='judge')[['rubric_minus_baseline','warning_minus_rubric']].agg(['mean','count']).to_csv(DATA/'instruction_by_judge.csv')
strict_framing=framing[framing.external & ~framing.tail_recovered]
strict_fw=strict_framing.pivot_table(index=['doc_id','case_id','judge','model'],columns='condition',values='clarity').reindex(columns=['baseline','rubric','warning']).dropna()
strict_effects=[]
for name,left,right in [('rubric_minus_baseline','rubric','baseline'),('warning_minus_rubric','warning','rubric')]:
 delta=(strict_fw[left]-strict_fw[right]).groupby(level='case_id').mean()
 strict_effects.append(dict(effect=name,strict_mean=delta.mean() if len(delta) else None,strict_cases=len(delta),strict_pairs=len(strict_fw)))
strict_effect_table=effect_table.merge(pd.DataFrame(strict_effects),on='effect')
strict_effect_table.to_csv(DATA/'strict_json_instruction_sensitivity.csv',index=False)
print('Strict JSON-only instruction sensitivity:');display(strict_effect_table.round(3))
````

### A real error that every judge caught, and a judge's own error

Claude Opus 5.5's web article for `F5-3` (document `4b18caca8fb62c3f`, 974 words, inside its range) runs a sensitivity test and gets it wrong:

> If the low scenario were 30% likely instead of 20%, with the remaining probability taken from the high case, expected gross value would fall to $2,820. Expected net value would then be **–$140**, and the expansion would fail Test 1.

With probabilities of 30%, 50% and 20% on scenario values of $1,200, $3,000 and $5,400, gross value is 0.30 × $1,200 + 0.50 × $3,000 + 0.20 × $5,400 = $2,940, and after the $2,960 cost the net is -$20, not -$140. The test still fails, so the recommendation stands, but the stated loss is wrong. All three judges found it. Claude Opus 5.5 wrote "so net is –$20, not –$140" and gave factual support 68; GPT-6 Astra gave the same correction and 79. GPT-6.1 Sol also scored factual support 80, but in a separate concern it made its own mistake: it said the break-even value is "approximately $26.79 ($2,960 divided by 112)". The division gives $26.43, which is what the article says. A judge can catch one error and make another in its explanation.

## 15. What the evidence supports

- When comparing writers with AI judges, give every writer the same judges and report each judge's scores alongside the average. Here the OpenAI judges scored every draft higher than the Claude judge, by 3 to 24 points, and dropping same-company judges turned that strictness gap into a ranking.
- Treat a rule that drops same-company judges as a source of bias in its own right unless the judges have been shown to use the scale the same way. Here, each judge company on its own put GPT-6 Astra first; the mixed panel put it last.
- Do not read a model judge's prose flag as a fault in the wording. Fresh judges seldom agreed that flagged passages had a wording problem, and when they did see a problem it was usually in the reasoning.
- A warning about "AI slop" made little or no difference to scores once the criteria were fixed (-0.31, interval -0.70 to +0.02).
- Check a judge's arithmetic as well as the writer's.

The uncertainty that most needs another experiment is which judge, if any, matches human readers: whether the Claude judge's lower scores and extra concerns mark documents that people find harder to follow.

## 16. Limitations

- Every judgment comes from a model. The reader and editor forms have no responses, so neither ranking, nor any concern, is shown to match what people find clear.
- There are only two judge companies and three judges, chosen by design. With two companies, an outside-company panel can never be the same for writers from both. A study with judges from several more companies might show a smaller or larger strictness spread.
- The cases are invented and give the writer every number and the rule. They cover six decision types with four cases each, and each case has one document type, so differences between document types also reflect different cases. Each writer wrote two drafts per case.
- The earlier passages come from only two cases, and the re-check asked different questions from the audit that produced the flags.
- Command-line products bring their own built-in instructions and default settings, so this compares configured products, not bare models. The study scores slide text, not slide layout.
- The protocol and cases were saved before any draft or score was collected, and the outside-company panel was the planned main analysis. Two things were added during the study: the parser repair for replies missing their closing brackets, and the common panel, which was not in the protocol. The examples and the breakdown by judge company were added after the results to explain them. This is an exploratory study; the record is in [`data/PROTOCOL.md`](data/PROTOCOL.md).

## 18. Future work and further reading

### Future work

- Collect the prepared reader and editor responses, and test whether the Claude judge's lower scores, or any judge's concerns, predict readers' accuracy and effort on the same documents. This is the test that could say which ranking in section 11 is closer to readers.
- Add judges from two or three more companies, every one scoring every draft, to measure how much judges differ in strictness and whether a model of each judge's scale (a per-judge offset) removes the ranking reversal.
- Give human editors verified, fact-preserving rewrites of flagged passages, with a proofread-only control, and randomly assign readers one version per case. This would measure whether fixing a flagged passage helps, which the re-check here cannot.
- Cross every case with all four document types, so that document type and decision type are no longer tied together.
- Ask the judges for schema-enforced replies through each tool's structured-output option and compare validity and scores with the plain JSON instructions used here, where 78 replies needed the bracket repair.
